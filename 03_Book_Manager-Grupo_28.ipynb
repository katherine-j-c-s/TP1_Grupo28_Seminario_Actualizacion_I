````python
#@title Ejercicio 02 — cargar_datos_desde_sql
from pathlib import Path
destino = Path(RUTA_SRC) / 'book_manager/migrations/migrations.py'
destino.parent.mkdir(parents=True, exist_ok=True)
destino.write_text(r'''import csv
import re
from pathlib import Path

from sqlalchemy import text

from book_manager.database.connection import ConexionDB

COLUMNAS_ENTERAS = {
    "id", "editorial_id", "genero_id", "anio", "paginas", "libro_id",
    "moneda_id", "cantidad", "tipo_cotizacion_id", "punto_reposicion",
}
COLUMNAS_DECIMALES = {"precio", "compra", "venta"}

ARCHIVOS_CSV = [
    {
        "archivo": "generos.csv",
        "tabla": "generos",
        "mapeo": {"id": "id", "nombre": "nombre", "descripcion": "descripcion"},
    },
    {
        "archivo": "editoriales.csv",
        "tabla": "editoriales",
        "mapeo": {
            "id": "id",
            "nombre": "nombre",
            "pais": "pais",
            "sitio_web": "sitio_web",
        },
    },
    {
        "archivo": "monedas.csv",
        "tabla": "monedas",
        "mapeo": {
            "id": "id",
            "codigo": "codigo",
            "nombre": "nombre",
            "simbolo": "simbolo",
        },
    },
    {
        "archivo": "tipos_cotizacion.csv",
        "tabla": "tipos_cotizacion",
        "mapeo": {"id": "id", "nombre": "nombre", "descripcion": "descripcion"},
    },
    {
        "archivo": "libros.csv",
        "tabla": "libros",
        "mapeo": {
            "id": "id",
            "isbn": "isbn",
            "titulo": "titulo",
            "autor": "autor",
            "editorial_id": "editorial_id",
            "genero_id": "genero_id",
            "anio": "anio",
            "paginas": "paginas",
        },
    },
    {
        "archivo": "precios.csv",
        "tabla": "precios_libro",
        "mapeo": {
            "id": "id",
            "libro_id": "libro_id",
            "moneda_id": "moneda_id",
            "importe": "precio",
        },
    },
    {
        "archivo": "stocks.csv",
        "tabla": "stock_libro",
        "mapeo": {
            "libro_id": "libro_id",
            "cantidad": "cantidad",
            "punto_reposicion": "punto_reposicion",
        },
    },
    {
        "archivo": "cotizaciones_dolar.csv",
        "tabla": "cotizaciones_dolar",
        "mapeo": {
            "tipo_id": "tipo_cotizacion_id",
            "fecha": "fecha",
            "compra": "compra",
            "venta": "venta",
        },
    },
]

OBLIGATORIAS = {
    "generos": {"id", "nombre"},
    "editoriales": {"id", "nombre"},
    "monedas": {"id", "codigo", "nombre", "simbolo"},
    "tipos_cotizacion": {"id", "nombre"},
    "libros": {"id", "isbn", "titulo", "autor", "editorial_id",
               "genero_id", "anio", "paginas"},
    "precios_libro": {"id", "libro_id", "moneda_id", "precio"},
    "stock_libro": {"libro_id", "cantidad"},
    "cotizaciones_dolar": {"tipo_cotizacion_id", "fecha", "compra", "venta"},
}


def _escapar_valor(valor):
    """Escapa comillas simples para generar sintaxis SQL valida.

    Args:
        valor (str): Valor de texto taken del CSV.

    Returns:
        str: Literal SQL entre comillas simples.
    """
    return "'" + valor.replace("'", "''") + "'"


def _formatear_valor(origen, destino):
    """Convierte un valor del CSV al literal SQL segun el tipo del destino.

    Args:
        origen (str): Valor crudo de la celda del CSV.
        destino (str): Nombre de la columna en el modelo.

    Returns:
        str: Literal SQL listo para el INSERT.
    """
    valor = (origen or "").strip()
    if valor == "":
        return "NULL"
    if destino in COLUMNAS_ENTERAS:
        return str(int(float(valor)))
    if destino in COLUMNAS_DECIMALES:
        return str(float(valor))
    return _escapar_valor(valor)


def migrar_datos(carpeta_csvs: str, carpeta_sqls: str, db: ConexionDB = None) -> None:
    """Migra los CSV a la base relacional y guarda los scripts SQL generados.

    Antes de insertar valida que cada CSV tenga las columnas del mapeo, de modo
    que una diferencia de nombres se reporte en lugar de terminar en un
    IntegrityError a mitad de la carga.

    Args:
        carpeta_csvs (str): Carpeta con los CSV de origen.
        carpeta_sqls (str): Carpeta donde se guardan los SQL generados.
        db (ConexionDB): Conexion a reutilizar. Si es None se crea una.

    Raises:
        FileNotFoundError: Si falta un CSV o le faltan columnas del mapeo.
    """
    path_csv = Path(carpeta_csvs)
    path_sql = Path(carpeta_sqls)
    path_sql.mkdir(parents=True, exist_ok=True)

    if db is None:
        db = ConexionDB()
    db.crear_tablas()

    problemas = []
    resumen = []

    for entrada in ARCHIVOS_CSV:
        archivo_csv = path_csv / entrada["archivo"]

        if not archivo_csv.exists():
            problemas.append(entrada["archivo"] + ": el archivo no existe")
            continue

        with open(archivo_csv, "r", encoding="utf-8-sig") as archivo:
            reader = csv.DictReader(archivo)
            encabezados = reader.fieldnames or []

            faltantes = [c for c in entrada["mapeo"] if c not in encabezados]
            if faltantes:
                problemas.append(
                    entrada["archivo"] + ": faltan columnas " + str(faltantes)
                    + " (encabezado real: " + str(encabezados) + ")"
                )
                continue

            destinos = list(entrada["mapeo"].values())
            obligatorias = OBLIGATORIAS.get(entrada["tabla"], set())

            lineas_sql = []
            descartadas = 0

            for numero, fila in enumerate(reader, start=2):
                vacias = [
                    destino
                    for origen, destino in entrada["mapeo"].items()
                    if destino in obligatorias
                    and not (fila.get(origen) or "").strip()
                ]
                if vacias:
                    descartadas += 1
                    problemas.append(
                        entrada["archivo"] + ": fila " + str(numero)
                        + " sin valores obligatorios en " + str(vacias)
                    )
                    continue

                valores = [
                    _formatear_valor(fila.get(origen), destino)
                    for origen, destino in entrada["mapeo"].items()
                ]
                lineas_sql.append(
                    "INSERT INTO " + entrada["tabla"]
                    + " (" + ", ".join(destinos) + ") VALUES ("
                    + ", ".join(valores) + ");\n"
                )

        archivo_sql = path_sql / (entrada["tabla"] + ".sql")
        with open(archivo_sql, "w", encoding="utf-8") as salida:
            salida.writelines(lineas_sql)

        with db.transaccion() as sesion:
            for sentencia in lineas_sql:
                if sentencia.strip():
                    sesion.execute(text(sentencia))

        resumen.append(
            "  " + entrada["archivo"] + " -> " + entrada["tabla"] + ": "
            + str(len(lineas_sql)) + " filas"
            + (" (" + str(descartadas) + " descartadas)" if descartadas else "")
        )

    print("\n".join(resumen))

    if problemas:
        raise FileNotFoundError(
            "Problemas en la migracion:\n"
            + "\n".join("  - " + p for p in problemas)
        )

    print("Migracion completada.")


def cargar_datos_desde_sql(
    carpeta_sqls: str,
    db: ConexionDB | None = None,
) -> dict[str, int]:
    """Carga los scripts .sql en la base.

    Si la tabla de un archivo ya tiene filas, no vuelve a insertarlas.
    Así se puede ejecutar de nuevo sin duplicar el catálogo.

    Args:
        carpeta_sqls (str): Carpeta con los .sql generados en el Sprint 2.
        db (ConexionDB): Conexión a reutilizar. Si es None se crea una.

    Returns:
        dict[str, int]: Cantidad de filas por archivo.

    Raises:
        FileNotFoundError: Si la carpeta no tiene scripts.
        ValueError: Si un script no empieza con INSERT INTO.
    """
    path_sql = Path(carpeta_sqls)
    archivos = sorted(path_sql.glob("*.sql"))
    if not archivos:
        raise FileNotFoundError(
            f"No hay scripts SQL para cargar en {carpeta_sqls}."
        )
    if db is None:
        db = ConexionDB()
    db.crear_tablas()
    resumen: dict[str, int] = {}
    for archivo in archivos:
        sentencias = [
            linea.strip()
            for linea in archivo.read_text(encoding="utf-8").splitlines()
            if linea.strip()
        ]
        if not sentencias:
            resumen[archivo.name] = 0
            continue
        tabla = _tabla_de_insert(sentencias[0], archivo.name)
        with db.engine.connect() as conexion:
            cantidad = conexion.execute(
                text(f"SELECT COUNT(*) FROM {tabla}")
            ).scalar()
        if cantidad:
            resumen[archivo.name] = int(cantidad)
            continue
        with db.transaccion() as sesion:
            for sentencia in sentencias:
                sesion.execute(text(sentencia))
        resumen[archivo.name] = len(sentencias)
    return resumen


def _tabla_de_insert(sentencia: str, archivo: str) -> str:
    """Obtiene el nombre de tabla de un INSERT INTO."""
    coincidencia = re.match(
        r"INSERT\s+INTO\s+([a-z_]+)\b",
        sentencia,
        flags=re.IGNORECASE,
    )
    if coincidencia is None:
        raise ValueError(f"{archivo} no empieza con INSERT INTO.")
    return coincidencia.group(1)
''', encoding='utf-8')
print('Vista —', destino.name, destino.stat().st_size, 'bytes')
````

````python
#@title Ejercicio 03 — scraper/loaders.py
from pathlib import Path
destino = Path(RUTA_SRC) / 'book_manager/scraper/loaders.py'
destino.parent.mkdir(parents=True, exist_ok=True)
destino.write_text(r'''"""Loaders del scraper de Cúspide."""

import re
from typing import Optional

from itemloaders import ItemLoader
from itemloaders.processors import MapCompose
from itemloaders.processors import TakeFirst

from book_manager.scraper.items import LibroWebItem

_METODOS = (
    "Transferencia",
    "Visa",
    "Mastercard",
    "American Express",
    "MercadoPago",
    "Mobbex",
    "RapiPago",
    "Pago Fácil",
)


def limpiar_precio(valor: object) -> Optional[float]:
    """Pasa un precio argentino ('38.999,00') a float.

    Args:
        valor (object): Texto o número leído de la página.

    Returns:
        Optional[float]: Precio, o None si no se pudo leer.
    """
    if valor is None:
        return None
    if isinstance(valor, (int, float)) and not isinstance(valor, bool):
        return float(valor)
    texto = str(valor).replace("\xa0", "").replace("$", "").strip()
    if texto == "":
        return None
    if "," in texto and "." in texto:
        texto = texto.replace(".", "").replace(",", ".")
    elif "," in texto:
        texto = texto.replace(",", ".")
    coincidencia = re.search(r"\d+(?:\.\d+)?", texto)
    if coincidencia is None:
        return None
    return float(coincidencia.group(0))


def _comprimir(texto: str) -> str:
    return " ".join(texto.split())


def formas_de_pago(html: str, precio: Optional[float]) -> str:
    """Arma 'medio: precio' con los medios publicados en la ficha.

    Args:
        html (str): HTML de la ficha.
        precio (Optional[float]): Precio de contado de esa ficha.

    Returns:
        str: Medios separados por ' | '.
    """
    bajo = html.lower()
    precio_txt = f"{precio:.2f}" if precio is not None else "s/p"
    partes = [
        f"{nombre}: {precio_txt}"
        for nombre in _METODOS
        if nombre.lower() in bajo
    ]
    dolar = re.search(r"U\$s\s*(?:<span[^>]*>)?\s*([0-9.,]+)", html, re.I)
    if dolar is not None:
        valor = limpiar_precio(dolar.group(1))
        if valor is not None:
            partes.append(f"Precio en dólares: {valor:.2f}")
    return " | ".join(partes)


class LibroCuspideLoader(ItemLoader):
    """Toma los campos de la ficha y se queda con el primer valor útil."""

    default_item_class = LibroWebItem
    default_output_processor = TakeFirst()
    titulo_web_in = MapCompose(str.strip)
    descripcion_in = MapCompose(_comprimir)
    url_imagen_in = MapCompose(str.strip)
    precio_web_in = MapCompose(limpiar_precio)
    formas_pago_in = MapCompose(str.strip)
''', encoding='utf-8')
print('Vista —', destino.name, destino.stat().st_size, 'bytes')
````

````python
#@title Ejercicio 03 — scraper/pipelines.py
from pathlib import Path
destino = Path(RUTA_SRC) / 'book_manager/scraper/pipelines.py'
destino.parent.mkdir(parents=True, exist_ok=True)
destino.write_text(r'''"""Pipelines del scraper de Cúspide."""

from datetime import datetime

from itemadapter import ItemAdapter
from scrapy.exceptions import DropItem

from book_manager.database.connection import ConexionDB
from book_manager.entities.entities import ExtraccionWeb
from book_manager.repositories.repositories import RepositorioExtraccionWeb
from book_manager.scraper.loaders import limpiar_precio


class LimpiezaPipeline:
    """Descarta fichas vacías y normaliza el precio y la descripción."""

    def process_item(self, item, spider):
        """Limpia un item antes de guardarlo.

        Args:
            item: Item producido por el loader.
            spider: Spider que lo generó.

        Returns:
            Item limpio.

        Raises:
            DropItem: Si no hay título ni precio.
        """
        datos = ItemAdapter(item)
        titulo = (datos.get("titulo_web") or "").strip()
        precio = datos.get("precio_web")
        if isinstance(precio, str):
            precio = limpiar_precio(precio)
            datos["precio_web"] = precio
        if titulo == "" and precio is None:
            raise DropItem("Resultado sin título ni precio.")
        datos["titulo_web"] = titulo or datos.get("titulo_buscado") or "Sin título"
        descripcion = datos.get("descripcion") or ""
        datos["descripcion"] = " ".join(str(descripcion).split())
        return item


class GuardarExtraccionPipeline:
    """Guarda cada ficha en la tabla extracciones_web."""

    def open_spider(self, spider) -> None:
        """Crea la tabla y vacía las extracciones de la corrida anterior."""
        import book_manager.models.models  # registra las tablas

        self._db = ConexionDB()
        self._db.crear_tablas()
        self._repositorio = RepositorioExtraccionWeb(self._db)
        self._repositorio.eliminar_todas()

    def process_item(self, item, spider):
        """Persiste el item.

        Args:
            item: Item ya limpio.
            spider: Spider que lo generó.

        Returns:
            El mismo item.
        """
        datos = ItemAdapter(item)
        self._repositorio.crear(
            ExtraccionWeb(
                id=0,
                libro_id=int(datos["libro_id"]),
                titulo_buscado=str(datos.get("titulo_buscado") or ""),
                titulo_web=str(datos.get("titulo_web") or ""),
                precio_web=datos.get("precio_web"),
                url_imagen=str(datos.get("url_imagen") or ""),
                formas_pago=str(datos.get("formas_pago") or ""),
                descripcion=str(datos.get("descripcion") or ""),
                url_producto=str(datos.get("url_producto") or ""),
                fecha_extraccion=datetime.now(),
                coincide=bool(datos.get("coincide")),
            )
        )
        return item
''', encoding='utf-8')
print('Vista —', destino.name, destino.stat().st_size, 'bytes')
````

````python
#@title Ejercicio 03 — scraper/spiders/cuspide.py
from pathlib import Path
destino = Path(RUTA_SRC) / 'book_manager/scraper/spiders/cuspide.py'
destino.parent.mkdir(parents=True, exist_ok=True)
destino.write_text(r'''"""Spider que busca en Cúspide solo los libros del catálogo."""

from urllib.parse import quote_plus

import scrapy

from book_manager.database.connection import ConexionDB
from book_manager.repositories.repositories import RepositorioLibro
from book_manager.scraper.items import LibroWebItem
from book_manager.scraper.loaders import LibroCuspideLoader
from book_manager.scraper.loaders import formas_de_pago
from book_manager.scraper.loaders import limpiar_precio


class CuspideSpider(scrapy.Spider):
    """Busca cada título del catálogo y toma los 10 primeros resultados."""

    name = "cuspide"
    allowed_domains = ["cuspide.com"]

    def __init__(self, limite: int = 0, *args, **kwargs) -> None:
        """Arma la lista de títulos desde la base.

        Args:
            limite (int): Si es mayor a cero, busca solo esa cantidad de libros.
        """
        super().__init__(*args, **kwargs)
        self.limite = int(limite or 0)
        self.libros = self._libros_del_catalogo()

    async def start(self):
        """Entrega las búsquedas iniciales. Scrapy 2.13+ llama a este método."""
        for pedido in self.start_requests():
            yield pedido

    def start_requests(self):
        """Una búsqueda por cada título del catálogo."""
        for libro in self.libros:
            consulta = quote_plus(libro["titulo"])
            url = f"https://cuspide.com/?s={consulta}&post_type=product"
            yield scrapy.Request(
                url,
                callback=self.parse_busqueda,
                cb_kwargs={
                    "libro_id": libro["id"],
                    "titulo_buscado": libro["titulo"],
                },
            )

    def parse_busqueda(self, response, libro_id: int, titulo_buscado: str):
        """Sigue los 10 primeros productos de la búsqueda."""
        vistos = []
        for href in response.css("div.product-small a::attr(href)").getall():
            if "/producto/" not in href:
                continue
            absoluto = response.urljoin(href.split("?")[0])
            if absoluto in vistos:
                continue
            vistos.append(absoluto)
            if len(vistos) == 10:
                break
        for url in vistos:
            yield response.follow(
                url,
                callback=self.parse_producto,
                cb_kwargs={
                    "libro_id": libro_id,
                    "titulo_buscado": titulo_buscado,
                },
            )

    def parse_producto(self, response, libro_id: int, titulo_buscado: str):
        """Extrae precio, imagen, formas de pago y descripción."""
        precio_txt = response.css(
            ".product-page-price .woocommerce-Price-amount bdi::text"
        ).get()
        precio = limpiar_precio(precio_txt)
        loader = LibroCuspideLoader(item=LibroWebItem(), selector=response.selector)
        loader.add_value("libro_id", libro_id)
        loader.add_value("titulo_buscado", titulo_buscado)
        loader.add_css("titulo_web", "h1.product-title::text")
        loader.add_css("titulo_web", "h1.entry-title::text")
        loader.add_css("titulo_web", "h1::text")
        loader.add_value("precio_web", precio)
        loader.add_css(
            "url_imagen",
            ".woocommerce-product-gallery__image a::attr(href)",
        )
        loader.add_css("descripcion", "p.livriz-text-synopsis::text")
        loader.add_value("url_producto", response.url)
        loader.add_value("formas_pago", formas_de_pago(response.text, precio))
        item = loader.load_item()
        item["coincide"] = _mismo_titulo(
            titulo_buscado,
            str(item.get("titulo_web") or ""),
        )
        yield item

    def _libros_del_catalogo(self) -> list:
        """Lee los títulos desde SQL. No busca libros fuera del catálogo."""
        import book_manager.models.models  # registra las tablas

        db = ConexionDB()
        db.crear_tablas()
        libros = RepositorioLibro(db).leer_todos()
        if self.limite > 0:
            libros = libros[: self.limite]
        return [{"id": libro.id, "titulo": libro.titulo} for libro in libros]


def _mismo_titulo(buscado: str, publicado: str) -> bool:
    """Compara títulos sin distinguir mayúsculas."""
    izquierdo = " ".join(buscado.split()).casefold()
    derecho = " ".join(publicado.split()).casefold()
    if izquierdo == "" or derecho == "":
        return False
    return izquierdo == derecho or izquierdo in derecho or derecho in izquierdo
''', encoding='utf-8')
print('Vista —', destino.name, destino.stat().st_size, 'bytes')
````

````python
#@title Ejercicio 04 — entidades de extracción
from pathlib import Path
destino = Path(RUTA_SRC) / 'book_manager/entities/entities.py'
destino.parent.mkdir(parents=True, exist_ok=True)
destino.write_text(r'''"""Entidades de Book Manager.

Cada atributo queda encapsulado: se guarda en un nombre privado y se
expone con una propiedad que valida el valor antes de aceptarlo.
"""

from datetime import date
from datetime import datetime


def _exigir_texto(valor: str, campo: str) -> str:
    """Devuelve el texto sin espacios laterales.

    Args:
        valor (str): Texto ingresado.
        campo (str): Nombre del campo, para el mensaje de error.

    Returns:
        str: Texto limpio.

    Raises:
        ValueError: Si el valor no es un texto o queda vacío.
    """
    if not isinstance(valor, str):
        raise ValueError(f"{campo} debe ser un texto.")
    limpio = valor.strip()
    if limpio == "":
        raise ValueError(f"{campo} no puede estar vacío.")
    return limpio


def _exigir_entero(valor: int, campo: str, permitir_cero: bool) -> int:
    """Valida un entero.

    Args:
        valor (int): Número a validar.
        campo (str): Nombre del campo.
        permitir_cero (bool): Permite el cero cuando es verdadero.

    Returns:
        int: El mismo entero, ya validado.

    Raises:
        ValueError: Si el valor no es un entero o está fuera de rango.
    """
    if type(valor) is not int:
        raise ValueError(f"{campo} debe ser un número entero.")
    if permitir_cero and valor < 0:
        raise ValueError(f"{campo} no puede ser negativo.")
    if not permitir_cero and valor <= 0:
        raise ValueError(f"{campo} debe ser mayor a cero.")
    return valor


def _exigir_instancia(valor: object, clase: type, campo: str) -> object:
    """Exige que el valor sea una instancia de la clase indicada.

    Args:
        valor (object): Objeto recibido.
        clase (type): Clase esperada.
        campo (str): Nombre del campo, para el mensaje de error.

    Returns:
        object: El mismo objeto, ya validado.

    Raises:
        ValueError: Si el valor no es una instancia de la clase.
    """
    if not isinstance(valor, clase):
        raise ValueError(f"{campo} debe ser un objeto {clase.__name__}.")
    return valor


def _exigir_importe(valor: float, campo: str) -> float:
    """Valida un importe positivo y lo redondea a dos decimales.

    Args:
        valor (float): Importe a validar.
        campo (str): Nombre del campo.

    Returns:
        float: Importe redondeado.

    Raises:
        ValueError: Si el valor no es numérico o no es positivo.
    """
    if isinstance(valor, bool) or not isinstance(valor, (int, float)):
        raise ValueError(f"{campo} debe ser un número.")
    importe = round(float(valor), 2)
    if importe <= 0:
        raise ValueError(f"{campo} debe ser mayor a cero.")
    return importe


class EntidadBase:
    """Entidad identificada por un id.

    El id 0 indica que el repositorio todavía no lo persistió.
    """

    def __init__(self, id: int) -> None:
        """Constructor.

        Args:
            id (int): Identificador. Cero si todavía no fue asignado.
        """
        self.__id = 0
        self.id = id

    @property
    def id(self) -> int:
        """Devuelve el identificador."""
        return self.__id

    @id.setter
    def id(self, valor: int) -> None:
        """Asigna el identificador.

        Args:
            valor (int): Nuevo identificador.

        Raises:
            ValueError: Si el valor no es un entero mayor o igual a cero.
        """
        self.__id = _exigir_entero(valor, "El id", permitir_cero=True)

    def __str__(self) -> str:
        return f"[{self.id}] {self.__class__.__name__}"


class Genero(EntidadBase):
    """Categoría literaria de un libro."""

    def __init__(self, id: int, nombre: str, descripcion: str) -> None:
        """Constructor.

        Args:
            id (int): Identificador.
            nombre (str): Nombre del género.
            descripcion (str): Descripción breve.
        """
        super().__init__(id)
        self.__nombre = ""
        self.__descripcion = ""
        self.nombre = nombre
        self.descripcion = descripcion

    @property
    def nombre(self) -> str:
        """Devuelve el nombre del género."""
        return self.__nombre

    @nombre.setter
    def nombre(self, valor: str) -> None:
        """Asigna el nombre del género."""
        self.__nombre = _exigir_texto(valor, "El nombre del género")

    @property
    def descripcion(self) -> str:
        """Devuelve la descripción del género."""
        return self.__descripcion

    @descripcion.setter
    def descripcion(self, valor: str) -> None:
        """Asigna la descripción del género."""
        self.__descripcion = _exigir_texto(valor, "La descripción del género")

    def __str__(self) -> str:
        return f"[{self.id}] {self.nombre}: {self.descripcion}"


class Editorial(EntidadBase):
    """Proveedor que distribuye los libros."""

    def __init__(self, id: int, nombre: str, pais: str, sitio_web: str) -> None:
        """Constructor.

        Args:
            id (int): Identificador.
            nombre (str): Nombre de la editorial.
            pais (str): País de origen.
            sitio_web (str): Sitio web.
        """
        super().__init__(id)
        self.__nombre = ""
        self.__pais = ""
        self.__sitio_web = ""
        self.nombre = nombre
        self.pais = pais
        self.sitio_web = sitio_web

    @property
    def nombre(self) -> str:
        """Devuelve el nombre de la editorial."""
        return self.__nombre

    @nombre.setter
    def nombre(self, valor: str) -> None:
        """Asigna el nombre de la editorial."""
        self.__nombre = _exigir_texto(valor, "El nombre de la editorial")

    @property
    def pais(self) -> str:
        """Devuelve el país de la editorial."""
        return self.__pais

    @pais.setter
    def pais(self, valor: str) -> None:
        """Asigna el país de la editorial."""
        self.__pais = _exigir_texto(valor, "El país")

    @property
    def sitio_web(self) -> str:
        """Devuelve el sitio web de la editorial."""
        return self.__sitio_web

    @sitio_web.setter
    def sitio_web(self, valor: str) -> None:
        """Asigna el sitio web de la editorial."""
        self.__sitio_web = _exigir_texto(valor, "El sitio web")

    def __str__(self) -> str:
        return f"[{self.id}] {self.nombre} ({self.pais}) - {self.sitio_web}"


class Moneda(EntidadBase):
    """Moneda en la que se puede expresar un precio."""

    def __init__(self, id: int, codigo: str, nombre: str, simbolo: str) -> None:
        """Constructor.

        Args:
            id (int): Identificador.
            codigo (str): Código de tres letras.
            nombre (str): Nombre de la moneda.
            simbolo (str): Símbolo.
        """
        super().__init__(id)
        self.__codigo = ""
        self.__nombre = ""
        self.__simbolo = ""
        self.codigo = codigo
        self.nombre = nombre
        self.simbolo = simbolo

    @property
    def codigo(self) -> str:
        """Devuelve el código de la moneda."""
        return self.__codigo

    @codigo.setter
    def codigo(self, valor: str) -> None:
        """Asigna el código de tres letras."""
        limpio = _exigir_texto(valor, "El código").upper()
        if len(limpio) != 3 or not limpio.isalpha():
            raise ValueError("El código de la moneda debe tener 3 letras.")
        self.__codigo = limpio

    @property
    def nombre(self) -> str:
        """Devuelve el nombre de la moneda."""
        return self.__nombre

    @nombre.setter
    def nombre(self, valor: str) -> None:
        """Asigna el nombre de la moneda."""
        self.__nombre = _exigir_texto(valor, "El nombre de la moneda")

    @property
    def simbolo(self) -> str:
        """Devuelve el símbolo de la moneda."""
        return self.__simbolo

    @simbolo.setter
    def simbolo(self, valor: str) -> None:
        """Asigna el símbolo de la moneda."""
        self.__simbolo = _exigir_texto(valor, "El símbolo")

    def __str__(self) -> str:
        return f"[{self.id}] {self.codigo} - {self.nombre} ({self.simbolo})"


class TipoCotizacion(EntidadBase):
    """Tipo de cotización del dólar."""

    def __init__(self, id: int, nombre: str, descripcion: str) -> None:
        """Constructor.

        Args:
            id (int): Identificador.
            nombre (str): Nombre del tipo. Por ejemplo, Blue o MEP.
            descripcion (str): Descripción breve.
        """
        super().__init__(id)
        self.__nombre = ""
        self.__descripcion = ""
        self.nombre = nombre
        self.descripcion = descripcion

    @property
    def nombre(self) -> str:
        """Devuelve el nombre del tipo de cotización."""
        return self.__nombre

    @nombre.setter
    def nombre(self, valor: str) -> None:
        """Asigna el nombre del tipo de cotización."""
        self.__nombre = _exigir_texto(valor, "El nombre del tipo de cotización")

    @property
    def descripcion(self) -> str:
        """Devuelve la descripción del tipo de cotización."""
        return self.__descripcion

    @descripcion.setter
    def descripcion(self, valor: str) -> None:
        """Asigna la descripción del tipo de cotización."""
        self.__descripcion = _exigir_texto(valor, "La descripción")

    def __str__(self) -> str:
        return f"[{self.id}] {self.nombre}: {self.descripcion}"


class Libro(EntidadBase):
    """Título del catálogo de la librería."""

    def __init__(
        self,
        id: int,
        isbn: str,
        titulo: str,
        autor: str,
        editorial: Editorial,
        genero: Genero,
        anio: int,
        paginas: int,
    ) -> None:
        """Constructor.

        Args:
            id (int): Identificador.
            isbn (str): Código de inventario del título.
            titulo (str): Título.
            autor (str): Autor.
            editorial (Editorial): Editorial del libro.
            genero (Genero): Género del libro.
            anio (int): Año de publicación.
            paginas (int): Cantidad de páginas.
        """
        super().__init__(id)
        self.__isbn = ""
        self.__titulo = ""
        self.__autor = ""
        self.__editorial = None
        self.__genero = None
        self.__anio = 0
        self.__paginas = 0
        self.isbn = isbn
        self.titulo = titulo
        self.autor = autor
        self.editorial = editorial
        self.genero = genero
        self.anio = anio
        self.paginas = paginas

    @property
    def isbn(self) -> str:
        """Devuelve el código de inventario."""
        return self.__isbn

    @isbn.setter
    def isbn(self, valor: str) -> None:
        """Asigna el código de inventario."""
        self.__isbn = _exigir_texto(valor, "El ISBN")

    @property
    def titulo(self) -> str:
        """Devuelve el título."""
        return self.__titulo

    @titulo.setter
    def titulo(self, valor: str) -> None:
        """Asigna el título."""
        self.__titulo = _exigir_texto(valor, "El título")

    @property
    def autor(self) -> str:
        """Devuelve el autor."""
        return self.__autor

    @autor.setter
    def autor(self, valor: str) -> None:
        """Asigna el autor."""
        self.__autor = _exigir_texto(valor, "El autor")

    @property
    def editorial(self) -> Editorial:
        """Devuelve la editorial del libro."""
        return self.__editorial

    @editorial.setter
    def editorial(self, valor: Editorial) -> None:
        """Asigna la editorial del libro."""
        self.__editorial = _exigir_instancia(valor, Editorial, "La editorial")

    @property
    def genero(self) -> Genero:
        """Devuelve el género del libro."""
        return self.__genero

    @genero.setter
    def genero(self, valor: Genero) -> None:
        """Asigna el género del libro."""
        self.__genero = _exigir_instancia(valor, Genero, "El género")

    @property
    def anio(self) -> int:
        """Devuelve el año de publicación."""
        return self.__anio

    @anio.setter
    def anio(self, valor: int) -> None:
        """Asigna el año de publicación."""
        anio = _exigir_entero(valor, "El año", permitir_cero=False)
        if anio < 1400 or anio > 2100:
            raise ValueError("El año debe estar entre 1400 y 2100.")
        self.__anio = anio

    @property
    def paginas(self) -> int:
        """Devuelve la cantidad de páginas."""
        return self.__paginas

    @paginas.setter
    def paginas(self, valor: int) -> None:
        """Asigna la cantidad de páginas."""
        self.__paginas = _exigir_entero(
            valor, "Las páginas", permitir_cero=False
        )

    def __str__(self) -> str:
        return (
            f"[{self.id}] {self.titulo} - {self.autor} "
            f"| {self.genero.nombre} | {self.editorial.nombre} "
            f"(ISBN {self.isbn}, {self.anio}, {self.paginas} pág.)"
        )


class Precio(EntidadBase):
    """Valor monetario de un libro en una moneda."""

    def __init__(
        self,
        id: int,
        libro: Libro,
        moneda: Moneda,
        importe: float,
    ) -> None:
        """Constructor.

        Args:
            id (int): Identificador.
            libro (Libro): Libro al que pertenece el precio.
            moneda (Moneda): Moneda del importe.
            importe (float): Importe.
        """
        super().__init__(id)
        self.__libro = None
        self.__moneda = None
        self.__importe = 0.0
        self.libro = libro
        self.moneda = moneda
        self.importe = importe

    @property
    def libro(self) -> Libro:
        """Devuelve el libro del precio."""
        return self.__libro

    @libro.setter
    def libro(self, valor: Libro) -> None:
        """Asigna el libro del precio."""
        self.__libro = _exigir_instancia(valor, Libro, "El libro")

    @property
    def moneda(self) -> Moneda:
        """Devuelve la moneda del precio."""
        return self.__moneda

    @moneda.setter
    def moneda(self, valor: Moneda) -> None:
        """Asigna la moneda del precio."""
        self.__moneda = _exigir_instancia(valor, Moneda, "La moneda")

    @property
    def importe(self) -> float:
        """Devuelve el importe."""
        return self.__importe

    @importe.setter
    def importe(self, valor: float) -> None:
        """Asigna el importe."""
        self.__importe = _exigir_importe(valor, "El importe")

    def __str__(self) -> str:
        return (
            f"[{self.id}] {self.libro.titulo} / {self.moneda.codigo}: "
            f"{self.importe:.2f}"
        )


class Stock:
    """Cantidad disponible de un libro.

    La clave es el libro: hay un único stock por título.
    """

    def __init__(
        self,
        libro: Libro,
        cantidad: int,
        punto_reposicion: int,
    ) -> None:
        """Constructor.

        Args:
            libro (Libro): Libro al que pertenece el stock.
            cantidad (int): Unidades disponibles.
            punto_reposicion (int): Cantidad a partir de la cual hay que reponer.
        """
        self.__libro = None
        self.__cantidad = 0
        self.__punto_reposicion = 0
        self.libro = libro
        self.cantidad = cantidad
        self.punto_reposicion = punto_reposicion

    @property
    def libro(self) -> Libro:
        """Devuelve el libro del stock."""
        return self.__libro

    @libro.setter
    def libro(self, valor: Libro) -> None:
        """Asigna el libro del stock."""
        self.__libro = _exigir_instancia(valor, Libro, "El libro")

    @property
    def cantidad(self) -> int:
        """Devuelve las unidades disponibles."""
        return self.__cantidad

    @cantidad.setter
    def cantidad(self, valor: int) -> None:
        """Asigna las unidades disponibles."""
        self.__cantidad = _exigir_entero(
            valor, "La cantidad", permitir_cero=True
        )

    @property
    def punto_reposicion(self) -> int:
        """Devuelve el punto de reposición."""
        return self.__punto_reposicion

    @punto_reposicion.setter
    def punto_reposicion(self, valor: int) -> None:
        """Asigna el punto de reposición."""
        self.__punto_reposicion = _exigir_entero(
            valor, "El punto de reposición", permitir_cero=True
        )

    def __str__(self) -> str:
        texto = "1 unidad" if self.cantidad == 1 else f"{self.cantidad} unidades"
        return (
            f"{self.libro.titulo}: {texto} "
            f"(reponer al llegar a {self.punto_reposicion})"
        )


class CotizacionDolar:
    """Cotización del dólar para un tipo y una fecha."""

    def __init__(
        self,
        tipo_cotizacion: TipoCotizacion,
        fecha: date,
        compra: float,
        venta: float,
    ) -> None:
        """Constructor.

        Args:
            tipo_cotizacion (TipoCotizacion): Tipo de dólar de la cotización.
            fecha (date): Fecha de la cotización.
            compra (float): Valor de compra.
            venta (float): Valor de venta.
        """
        self.__tipo_cotizacion = None
        self.__fecha = date(2026, 10, 1)
        self.__compra = 0.0
        self.__venta = 0.0
        self.tipo_cotizacion = tipo_cotizacion
        self.fecha = fecha
        self.compra = compra
        self.venta = venta

    @property
    def tipo_cotizacion(self) -> TipoCotizacion:
        """Devuelve el tipo de cotización."""
        return self.__tipo_cotizacion

    @tipo_cotizacion.setter
    def tipo_cotizacion(self, valor: TipoCotizacion) -> None:
        """Asigna el tipo de cotización."""
        self.__tipo_cotizacion = _exigir_instancia(
            valor, TipoCotizacion, "El tipo de cotización"
        )

    @property
    def fecha(self) -> date:
        """Devuelve la fecha de la cotización."""
        return self.__fecha

    @fecha.setter
    def fecha(self, valor: date) -> None:
        """Asigna la fecha de la cotización."""
        if type(valor) is not date:
            raise ValueError("La fecha debe ser un datetime.date.")
        self.__fecha = valor

    @property
    def compra(self) -> float:
        """Devuelve el valor de compra."""
        return self.__compra

    @compra.setter
    def compra(self, valor: float) -> None:
        """Asigna el valor de compra."""
        compra = _exigir_importe(valor, "La compra")
        if self.__venta and compra > self.__venta:
            raise ValueError("La compra no puede superar a la venta.")
        self.__compra = compra

    @property
    def venta(self) -> float:
        """Devuelve el valor de venta."""
        return self.__venta

    @venta.setter
    def venta(self, valor: float) -> None:
        """Asigna el valor de venta."""
        venta = _exigir_importe(valor, "La venta")
        if self.__compra and venta < self.__compra:
            raise ValueError("La venta no puede ser menor que la compra.")
        self.__venta = venta

    def __str__(self) -> str:
        return (
            f"{self.tipo_cotizacion.nombre} @ {self.fecha.isoformat()}: "
            f"compra {self.compra:.2f} / venta {self.venta:.2f}"
        )


class ExtraccionWeb:
    """Dato obtenido de Cúspide para un libro del catálogo."""

    def __init__(
        self,
        id: int,
        libro_id: int,
        titulo_buscado: str,
        titulo_web: str,
        precio_web: float | None,
        url_imagen: str,
        formas_pago: str,
        descripcion: str,
        url_producto: str,
        fecha_extraccion: datetime,
        coincide: bool,
    ) -> None:
        self.__id = id
        self.__libro_id = libro_id
        self.__titulo_buscado = titulo_buscado
        self.__titulo_web = titulo_web
        self.__precio_web = precio_web
        self.__url_imagen = url_imagen
        self.__formas_pago = formas_pago
        self.__descripcion = descripcion
        self.__url_producto = url_producto
        self.__fecha_extraccion = fecha_extraccion
        self.__coincide = coincide

    @property
    def id(self) -> int:
        """Devuelve el id."""
        return self.__id

    @property
    def libro_id(self) -> int:
        """Devuelve el id del libro del catálogo."""
        return self.__libro_id

    @property
    def titulo_buscado(self) -> str:
        """Devuelve el título con el que se buscó."""
        return self.__titulo_buscado

    @property
    def titulo_web(self) -> str:
        """Devuelve el título publicado en la web."""
        return self.__titulo_web

    @property
    def precio_web(self) -> float | None:
        """Devuelve el precio publicado, si se pudo leer."""
        return self.__precio_web

    @property
    def url_imagen(self) -> str:
        """Devuelve la URL de la imagen."""
        return self.__url_imagen

    @property
    def formas_pago(self) -> str:
        """Devuelve las formas de pago y su precio."""
        return self.__formas_pago

    @property
    def descripcion(self) -> str:
        """Devuelve la descripción detallada."""
        return self.__descripcion

    @property
    def url_producto(self) -> str:
        """Devuelve la URL de la ficha."""
        return self.__url_producto

    @property
    def fecha_extraccion(self) -> datetime:
        """Devuelve la fecha de extracción."""
        return self.__fecha_extraccion

    @property
    def coincide(self) -> bool:
        """Indica si el título web corresponde al libro buscado."""
        return self.__coincide

    def __str__(self) -> str:
        precio = f"{self.precio_web:.2f}" if self.precio_web is not None else "s/p"
        return f"{self.titulo_web} | web {precio} | {self.url_producto}"


class Auditoria:
    """Una fila del historial de auditoría."""

    def __init__(
        self,
        id: int,
        accion: str,
        fecha: datetime,
        detalles: str,
    ) -> None:
        self.__id = id
        self.__accion = accion
        self.__fecha = fecha
        self.__detalles = detalles

    @property
    def id(self) -> int:
        """Devuelve el id."""
        return self.__id

    @property
    def accion(self) -> str:
        """Devuelve la acción auditada."""
        return self.__accion

    @property
    def fecha(self) -> datetime:
        """Devuelve la fecha de la operación."""
        return self.__fecha

    @property
    def detalles(self) -> str:
        """Devuelve el detalle de la operación."""
        return self.__detalles

    def __str__(self) -> str:
        marca = self.fecha.strftime("%Y-%m-%d %H:%M:%S")
        return f"{marca} | {self.accion} | {self.detalles}"
''', encoding='utf-8')
print('Vista —', destino.name, destino.stat().st_size, 'bytes')
````

````python
#@title Ejercicio 04 — repositorio de extracciones
from pathlib import Path
destino = Path(RUTA_SRC) / 'book_manager/repositories/repositories.py'
destino.parent.mkdir(parents=True, exist_ok=True)
destino.write_text(r'''"""Repositorios SQLAlchemy de Book Manager.

Cada clase concreta implementa el CRUD de una entidad y hereda la
interfaz del Sprint 1, así la firma queda obligada. Las relaciones
se guardan como objetos: el id de la clave foránea se toma del objeto
relacionado al persistir.
"""

from abc import ABC
from abc import abstractmethod
from datetime import date
from typing import Generic
from typing import List
from typing import Optional
from typing import TypeVar

from sqlalchemy.orm import joinedload

from book_manager.database.connection import ConexionDB
from book_manager.entities.entities import Auditoria
from book_manager.entities.entities import CotizacionDolar
from book_manager.entities.entities import Editorial
from book_manager.entities.entities import ExtraccionWeb
from book_manager.entities.entities import Genero
from book_manager.entities.entities import Libro
from book_manager.entities.entities import Moneda
from book_manager.entities.entities import Precio
from book_manager.entities.entities import Stock
from book_manager.entities.entities import TipoCotizacion
from book_manager.models.models import AuditoriaModel
from book_manager.models.models import CotizacionDolarModel
from book_manager.models.models import EditorialModel
from book_manager.models.models import ExtraccionWebModel
from book_manager.models.models import GeneroModel
from book_manager.models.models import LibroModel
from book_manager.models.models import MonedaModel
from book_manager.models.models import PrecioLibroModel
from book_manager.models.models import StockLibroModel
from book_manager.models.models import TipoCotizacionModel

T = TypeVar("T")


class IRepositorio(ABC, Generic[T]):
    """Interfaz del CRUD de una entidad."""

    @abstractmethod
    def crear(self, entidad: T) -> T:
        """Crea una entidad.

        Args:
            entidad (T): Entidad a crear.

        Returns:
            T: Entidad creada.

        Raises:
            ValueError: Si ya existe una entidad con el mismo id.
        """

    @abstractmethod
    def leer_por_id(self, id: int) -> Optional[T]:
        """Lee una entidad por id.

        Args:
            id (int): Id de la entidad.

        Returns:
            Optional[T]: La entidad, o None si no está.
        """

    @abstractmethod
    def leer_todos(self) -> List[T]:
        """Lee todas las entidades.

        Returns:
            List[T]: Entidades cargadas.
        """

    @abstractmethod
    def actualizar(self, entidad: T) -> T:
        """Actualiza una entidad existente.

        Args:
            entidad (T): Entidad con el id a actualizar.

        Returns:
            T: Entidad actualizada.

        Raises:
            ValueError: Si no se encuentra la entidad.
        """

    @abstractmethod
    def eliminar(self, id: int) -> bool:
        """Elimina una entidad por id.

        Args:
            id (int): Id de la entidad a eliminar.

        Returns:
            bool: True si se eliminó, False si no estaba.
        """


class IRepositorioStock(ABC):
    """Interfaz del repositorio de stock, identificado por libro."""

    @abstractmethod
    def crear(self, stock: Stock) -> Stock:
        """Crea el stock de un libro.

        Args:
            stock (Stock): Stock a crear.

        Returns:
            Stock: Stock creado.

        Raises:
            ValueError: Si ese libro ya tiene stock.
        """

    @abstractmethod
    def leer_por_libro(self, libro_id: int) -> Optional[Stock]:
        """Lee el stock de un libro.

        Args:
            libro_id (int): Id del libro.

        Returns:
            Optional[Stock]: El stock, o None si no está.
        """

    @abstractmethod
    def leer_todos(self) -> List[Stock]:
        """Lee todos los registros de stock.

        Returns:
            List[Stock]: Stocks cargados.
        """

    @abstractmethod
    def actualizar(self, stock: Stock) -> Stock:
        """Actualiza el stock de un libro.

        Args:
            stock (Stock): Stock a actualizar.

        Returns:
            Stock: Stock actualizado.

        Raises:
            ValueError: Si no se encuentra el stock.
        """

    @abstractmethod
    def eliminar(self, libro_id: int) -> bool:
        """Elimina el stock de un libro.

        Args:
            libro_id (int): Id del libro.

        Returns:
            bool: True si se eliminó, False si no estaba.
        """


class IRepositorioCotizacionDolar(ABC):
    """Interfaz del repositorio de cotizaciones, identificado por tipo y fecha."""

    @abstractmethod
    def crear(self, cotizacion: CotizacionDolar) -> CotizacionDolar:
        """Crea una cotización.

        Args:
            cotizacion (CotizacionDolar): Cotización a crear.

        Returns:
            CotizacionDolar: Cotización creada.

        Raises:
            ValueError: Si ya existe una cotización para ese tipo y fecha.
        """

    @abstractmethod
    def leer_por_tipo_y_fecha(
        self,
        tipo_id: int,
        fecha: date,
    ) -> Optional[CotizacionDolar]:
        """Lee una cotización por tipo y fecha.

        Args:
            tipo_id (int): Id del tipo de cotización.
            fecha (date): Fecha de la cotización.

        Returns:
            Optional[CotizacionDolar]: La cotización, o None si no está.
        """

    @abstractmethod
    def leer_historico_por_tipo(self, tipo_id: int) -> List[CotizacionDolar]:
        """Lee el histórico de un tipo de cotización.

        Args:
            tipo_id (int): Id del tipo de cotización.

        Returns:
            List[CotizacionDolar]: Cotizaciones de ese tipo, por fecha.
        """

    @abstractmethod
    def leer_todas(self) -> List[CotizacionDolar]:
        """Lee todas las cotizaciones.

        Returns:
            List[CotizacionDolar]: Cotizaciones cargadas.
        """

    @abstractmethod
    def actualizar(self, cotizacion: CotizacionDolar) -> CotizacionDolar:
        """Actualiza una cotización existente.

        Args:
            cotizacion (CotizacionDolar): Cotización a actualizar.

        Returns:
            CotizacionDolar: Cotización actualizada.

        Raises:
            ValueError: Si no se encuentra la cotización.
        """

    @abstractmethod
    def eliminar(self, tipo_id: int, fecha: date) -> bool:
        """Elimina una cotización por tipo y fecha.

        Args:
            tipo_id (int): Id del tipo de cotización.
            fecha (date): Fecha de la cotización.

        Returns:
            bool: True si se eliminó, False si no estaba.
        """


def a_genero(modelo: GeneroModel) -> Genero:
    """Convierte un modelo de género en la entidad de dominio."""
    return Genero(
        id=modelo.id,
        nombre=modelo.nombre,
        descripcion=modelo.descripcion or "Sin descripción",
    )


def a_editorial(modelo: EditorialModel) -> Editorial:
    """Convierte un modelo de editorial en la entidad de dominio."""
    return Editorial(
        id=modelo.id,
        nombre=modelo.nombre,
        pais=modelo.pais or "Sin país",
        sitio_web=modelo.sitio_web or "https://cuspide.com",
    )


def a_moneda(modelo: MonedaModel) -> Moneda:
    """Convierte un modelo de moneda en la entidad de dominio."""
    return Moneda(
        id=modelo.id,
        codigo=modelo.codigo,
        nombre=modelo.nombre,
        simbolo=modelo.simbolo,
    )


def a_tipo(modelo: TipoCotizacionModel) -> TipoCotizacion:
    """Convierte un modelo de tipo de cotización en la entidad de dominio."""
    return TipoCotizacion(
        id=modelo.id,
        nombre=modelo.nombre,
        descripcion=modelo.descripcion or "Sin descripción",
    )


def a_libro(modelo: LibroModel) -> Libro:
    """Convierte un modelo de libro y adjunta género y editorial."""
    return Libro(
        id=modelo.id,
        isbn=modelo.isbn,
        titulo=modelo.titulo,
        autor=modelo.autor,
        editorial=a_editorial(modelo.editorial),
        genero=a_genero(modelo.genero),
        anio=modelo.anio,
        paginas=modelo.paginas,
    )


def a_precio(modelo: PrecioLibroModel) -> Precio:
    """Convierte un modelo de precio y adjunta libro y moneda."""
    return Precio(
        id=modelo.id,
        libro=a_libro(modelo.libro),
        moneda=a_moneda(modelo.moneda),
        importe=modelo.precio,
    )


def a_stock(modelo: StockLibroModel) -> Stock:
    """Convierte un modelo de stock y adjunta el libro."""
    return Stock(
        libro=a_libro(modelo.libro),
        cantidad=modelo.cantidad,
        punto_reposicion=modelo.punto_reposicion or 0,
    )


def a_cotizacion(modelo: CotizacionDolarModel) -> CotizacionDolar:
    """Convierte un modelo de cotización y adjunta el tipo."""
    return CotizacionDolar(
        tipo_cotizacion=a_tipo(modelo.tipo_cotizacion),
        fecha=modelo.fecha,
        compra=modelo.compra,
        venta=modelo.venta,
    )


def _opciones_libro():
    """Carga género y editorial junto con el libro."""
    return (
        joinedload(LibroModel.editorial),
        joinedload(LibroModel.genero),
    )


class RepositorioGenero(IRepositorio[Genero]):
    """CRUD de géneros sobre la tabla generos."""

    def __init__(self, db: ConexionDB) -> None:
        self._db = db

    def crear(self, entidad: Genero) -> Genero:
        """Inserta un género. Si el id es 0, lo asigna la base."""
        with self._db.transaccion() as session:
            if entidad.id and session.get(GeneroModel, entidad.id) is not None:
                raise ValueError("Ya existe una entidad con el mismo ID.")
            modelo = GeneroModel(
                nombre=entidad.nombre,
                descripcion=entidad.descripcion,
            )
            if entidad.id:
                modelo.id = entidad.id
            session.add(modelo)
            session.flush()
            entidad.id = modelo.id
            return a_genero(modelo)

    def leer_por_id(self, id: int) -> Optional[Genero]:
        """Lee un género por id."""
        with self._db.transaccion() as session:
            modelo = session.get(GeneroModel, id)
            if modelo is None:
                return None
            return a_genero(modelo)

    def leer_todos(self) -> List[Genero]:
        """Lee todos los géneros ordenados por id."""
        with self._db.transaccion() as session:
            filas = session.query(GeneroModel).order_by(GeneroModel.id).all()
            return [a_genero(fila) for fila in filas]

    def actualizar(self, entidad: Genero) -> Genero:
        """Actualiza un género existente."""
        with self._db.transaccion() as session:
            modelo = session.get(GeneroModel, entidad.id)
            if modelo is None:
                raise ValueError("No se encuentra la entidad para actualizar.")
            modelo.nombre = entidad.nombre
            modelo.descripcion = entidad.descripcion
            session.flush()
            return a_genero(modelo)

    def eliminar(self, id: int) -> bool:
        """Elimina un género por id."""
        with self._db.transaccion() as session:
            modelo = session.get(GeneroModel, id)
            if modelo is None:
                return False
            session.delete(modelo)
            return True


class RepositorioEditorial(IRepositorio[Editorial]):
    """CRUD de editoriales."""

    def __init__(self, db: ConexionDB) -> None:
        self._db = db

    def crear(self, entidad: Editorial) -> Editorial:
        """Inserta una editorial."""
        with self._db.transaccion() as session:
            if entidad.id and session.get(EditorialModel, entidad.id) is not None:
                raise ValueError("Ya existe una entidad con el mismo ID.")
            modelo = EditorialModel(
                nombre=entidad.nombre,
                pais=entidad.pais,
                sitio_web=entidad.sitio_web,
            )
            if entidad.id:
                modelo.id = entidad.id
            session.add(modelo)
            session.flush()
            entidad.id = modelo.id
            return a_editorial(modelo)

    def leer_por_id(self, id: int) -> Optional[Editorial]:
        """Lee una editorial por id."""
        with self._db.transaccion() as session:
            modelo = session.get(EditorialModel, id)
            if modelo is None:
                return None
            return a_editorial(modelo)

    def leer_todos(self) -> List[Editorial]:
        """Lee todas las editoriales."""
        with self._db.transaccion() as session:
            filas = session.query(EditorialModel).order_by(EditorialModel.id).all()
            return [a_editorial(fila) for fila in filas]

    def actualizar(self, entidad: Editorial) -> Editorial:
        """Actualiza una editorial."""
        with self._db.transaccion() as session:
            modelo = session.get(EditorialModel, entidad.id)
            if modelo is None:
                raise ValueError("No se encuentra la entidad para actualizar.")
            modelo.nombre = entidad.nombre
            modelo.pais = entidad.pais
            modelo.sitio_web = entidad.sitio_web
            session.flush()
            return a_editorial(modelo)

    def eliminar(self, id: int) -> bool:
        """Elimina una editorial por id."""
        with self._db.transaccion() as session:
            modelo = session.get(EditorialModel, id)
            if modelo is None:
                return False
            session.delete(modelo)
            return True


class RepositorioMoneda(IRepositorio[Moneda]):
    """CRUD de monedas."""

    def __init__(self, db: ConexionDB) -> None:
        self._db = db

    def crear(self, entidad: Moneda) -> Moneda:
        """Inserta una moneda."""
        with self._db.transaccion() as session:
            if entidad.id and session.get(MonedaModel, entidad.id) is not None:
                raise ValueError("Ya existe una entidad con el mismo ID.")
            modelo = MonedaModel(
                codigo=entidad.codigo,
                nombre=entidad.nombre,
                simbolo=entidad.simbolo,
            )
            if entidad.id:
                modelo.id = entidad.id
            session.add(modelo)
            session.flush()
            entidad.id = modelo.id
            return a_moneda(modelo)

    def leer_por_id(self, id: int) -> Optional[Moneda]:
        """Lee una moneda por id."""
        with self._db.transaccion() as session:
            modelo = session.get(MonedaModel, id)
            if modelo is None:
                return None
            return a_moneda(modelo)

    def leer_todos(self) -> List[Moneda]:
        """Lee todas las monedas."""
        with self._db.transaccion() as session:
            filas = session.query(MonedaModel).order_by(MonedaModel.id).all()
            return [a_moneda(fila) for fila in filas]

    def actualizar(self, entidad: Moneda) -> Moneda:
        """Actualiza una moneda."""
        with self._db.transaccion() as session:
            modelo = session.get(MonedaModel, entidad.id)
            if modelo is None:
                raise ValueError("No se encuentra la entidad para actualizar.")
            modelo.codigo = entidad.codigo
            modelo.nombre = entidad.nombre
            modelo.simbolo = entidad.simbolo
            session.flush()
            return a_moneda(modelo)

    def eliminar(self, id: int) -> bool:
        """Elimina una moneda por id."""
        with self._db.transaccion() as session:
            modelo = session.get(MonedaModel, id)
            if modelo is None:
                return False
            session.delete(modelo)
            return True


class RepositorioTipoCotizacion(IRepositorio[TipoCotizacion]):
    """CRUD de tipos de cotización."""

    def __init__(self, db: ConexionDB) -> None:
        self._db = db

    def crear(self, entidad: TipoCotizacion) -> TipoCotizacion:
        """Inserta un tipo de cotización."""
        with self._db.transaccion() as session:
            if entidad.id and session.get(TipoCotizacionModel, entidad.id) is not None:
                raise ValueError("Ya existe una entidad con el mismo ID.")
            modelo = TipoCotizacionModel(
                nombre=entidad.nombre,
                descripcion=entidad.descripcion,
            )
            if entidad.id:
                modelo.id = entidad.id
            session.add(modelo)
            session.flush()
            entidad.id = modelo.id
            return a_tipo(modelo)

    def leer_por_id(self, id: int) -> Optional[TipoCotizacion]:
        """Lee un tipo de cotización por id."""
        with self._db.transaccion() as session:
            modelo = session.get(TipoCotizacionModel, id)
            if modelo is None:
                return None
            return a_tipo(modelo)

    def leer_todos(self) -> List[TipoCotizacion]:
        """Lee todos los tipos de cotización."""
        with self._db.transaccion() as session:
            filas = (
                session.query(TipoCotizacionModel)
                .order_by(TipoCotizacionModel.id)
                .all()
            )
            return [a_tipo(fila) for fila in filas]

    def actualizar(self, entidad: TipoCotizacion) -> TipoCotizacion:
        """Actualiza un tipo de cotización."""
        with self._db.transaccion() as session:
            modelo = session.get(TipoCotizacionModel, entidad.id)
            if modelo is None:
                raise ValueError("No se encuentra la entidad para actualizar.")
            modelo.nombre = entidad.nombre
            modelo.descripcion = entidad.descripcion
            session.flush()
            return a_tipo(modelo)

    def eliminar(self, id: int) -> bool:
        """Elimina un tipo de cotización por id."""
        with self._db.transaccion() as session:
            modelo = session.get(TipoCotizacionModel, id)
            if modelo is None:
                return False
            session.delete(modelo)
            return True


class RepositorioLibro(IRepositorio[Libro]):
    """CRUD de libros. Género y editorial se persisten por el objeto relacionado."""

    def __init__(self, db: ConexionDB) -> None:
        self._db = db

    def crear(self, entidad: Libro) -> Libro:
        """Inserta un libro."""
        with self._db.transaccion() as session:
            if entidad.id and session.get(LibroModel, entidad.id) is not None:
                raise ValueError("Ya existe una entidad con el mismo ID.")
            modelo = LibroModel(
                isbn=entidad.isbn,
                titulo=entidad.titulo,
                autor=entidad.autor,
                editorial_id=entidad.editorial.id,
                genero_id=entidad.genero.id,
                anio=entidad.anio,
                paginas=entidad.paginas,
            )
            if entidad.id:
                modelo.id = entidad.id
            session.add(modelo)
            session.flush()
            entidad.id = modelo.id
            cargado = (
                session.query(LibroModel)
                .options(*_opciones_libro())
                .filter(LibroModel.id == modelo.id)
                .one()
            )
            return a_libro(cargado)

    def leer_por_id(self, id: int) -> Optional[Libro]:
        """Lee un libro por id, con su género y su editorial."""
        with self._db.transaccion() as session:
            modelo = (
                session.query(LibroModel)
                .options(*_opciones_libro())
                .filter(LibroModel.id == id)
                .one_or_none()
            )
            if modelo is None:
                return None
            return a_libro(modelo)

    def leer_todos(self) -> List[Libro]:
        """Lee todos los libros."""
        with self._db.transaccion() as session:
            filas = (
                session.query(LibroModel)
                .options(*_opciones_libro())
                .order_by(LibroModel.id)
                .all()
            )
            return [a_libro(fila) for fila in filas]

    def actualizar(self, entidad: Libro) -> Libro:
        """Actualiza un libro."""
        with self._db.transaccion() as session:
            modelo = session.get(LibroModel, entidad.id)
            if modelo is None:
                raise ValueError("No se encuentra la entidad para actualizar.")
            modelo.isbn = entidad.isbn
            modelo.titulo = entidad.titulo
            modelo.autor = entidad.autor
            modelo.editorial_id = entidad.editorial.id
            modelo.genero_id = entidad.genero.id
            modelo.anio = entidad.anio
            modelo.paginas = entidad.paginas
            session.flush()
            cargado = (
                session.query(LibroModel)
                .options(*_opciones_libro())
                .filter(LibroModel.id == modelo.id)
                .one()
            )
            return a_libro(cargado)

    def eliminar(self, id: int) -> bool:
        """Elimina un libro por id."""
        with self._db.transaccion() as session:
            modelo = session.get(LibroModel, id)
            if modelo is None:
                return False
            session.delete(modelo)
            return True


class RepositorioPrecio(IRepositorio[Precio]):
    """CRUD de precios. El libro y la moneda son objetos."""

    def __init__(self, db: ConexionDB) -> None:
        self._db = db

    def _cargar(self, session, id: int) -> PrecioLibroModel:
        return (
            session.query(PrecioLibroModel)
            .options(
                joinedload(PrecioLibroModel.moneda),
                joinedload(PrecioLibroModel.libro).joinedload(LibroModel.editorial),
                joinedload(PrecioLibroModel.libro).joinedload(LibroModel.genero),
            )
            .filter(PrecioLibroModel.id == id)
            .one()
        )

    def crear(self, entidad: Precio) -> Precio:
        """Inserta un precio."""
        with self._db.transaccion() as session:
            if entidad.id and session.get(PrecioLibroModel, entidad.id) is not None:
                raise ValueError("Ya existe una entidad con el mismo ID.")
            modelo = PrecioLibroModel(
                libro_id=entidad.libro.id,
                moneda_id=entidad.moneda.id,
                precio=entidad.importe,
            )
            if entidad.id:
                modelo.id = entidad.id
            session.add(modelo)
            session.flush()
            entidad.id = modelo.id
            return a_precio(self._cargar(session, modelo.id))

    def leer_por_id(self, id: int) -> Optional[Precio]:
        """Lee un precio por id."""
        with self._db.transaccion() as session:
            if session.get(PrecioLibroModel, id) is None:
                return None
            return a_precio(self._cargar(session, id))

    def leer_todos(self) -> List[Precio]:
        """Lee todos los precios."""
        with self._db.transaccion() as session:
            filas = (
                session.query(PrecioLibroModel)
                .options(
                    joinedload(PrecioLibroModel.moneda),
                    joinedload(PrecioLibroModel.libro).joinedload(LibroModel.editorial),
                    joinedload(PrecioLibroModel.libro).joinedload(LibroModel.genero),
                )
                .order_by(PrecioLibroModel.id)
                .all()
            )
            return [a_precio(fila) for fila in filas]

    def actualizar(self, entidad: Precio) -> Precio:
        """Actualiza un precio."""
        with self._db.transaccion() as session:
            modelo = session.get(PrecioLibroModel, entidad.id)
            if modelo is None:
                raise ValueError("No se encuentra la entidad para actualizar.")
            modelo.libro_id = entidad.libro.id
            modelo.moneda_id = entidad.moneda.id
            modelo.precio = entidad.importe
            session.flush()
            return a_precio(self._cargar(session, modelo.id))

    def eliminar(self, id: int) -> bool:
        """Elimina un precio por id."""
        with self._db.transaccion() as session:
            modelo = session.get(PrecioLibroModel, id)
            if modelo is None:
                return False
            session.delete(modelo)
            return True


class RepositorioStock(IRepositorioStock):
    """CRUD de stock. La clave de negocio es el libro."""

    def __init__(self, db: ConexionDB) -> None:
        self._db = db

    def _opciones(self):
        return (
            joinedload(StockLibroModel.libro).joinedload(LibroModel.editorial),
            joinedload(StockLibroModel.libro).joinedload(LibroModel.genero),
        )

    def crear(self, stock: Stock) -> Stock:
        """Inserta el stock de un libro."""
        with self._db.transaccion() as session:
            existente = (
                session.query(StockLibroModel)
                .filter(StockLibroModel.libro_id == stock.libro.id)
                .one_or_none()
            )
            if existente is not None:
                raise ValueError("Ya existe un registro de stock para el mismo libro.")
            modelo = StockLibroModel(
                libro_id=stock.libro.id,
                cantidad=stock.cantidad,
                punto_reposicion=stock.punto_reposicion,
            )
            session.add(modelo)
            session.flush()
            cargado = (
                session.query(StockLibroModel)
                .options(*self._opciones())
                .filter(StockLibroModel.id == modelo.id)
                .one()
            )
            return a_stock(cargado)

    def leer_por_libro(self, libro_id: int) -> Optional[Stock]:
        """Lee el stock de un libro."""
        with self._db.transaccion() as session:
            modelo = (
                session.query(StockLibroModel)
                .options(*self._opciones())
                .filter(StockLibroModel.libro_id == libro_id)
                .one_or_none()
            )
            if modelo is None:
                return None
            return a_stock(modelo)

    def leer_todos(self) -> List[Stock]:
        """Lee todos los stocks."""
        with self._db.transaccion() as session:
            filas = (
                session.query(StockLibroModel)
                .options(*self._opciones())
                .order_by(StockLibroModel.libro_id)
                .all()
            )
            return [a_stock(fila) for fila in filas]

    def actualizar(self, stock: Stock) -> Stock:
        """Actualiza el stock de un libro."""
        with self._db.transaccion() as session:
            modelo = (
                session.query(StockLibroModel)
                .filter(StockLibroModel.libro_id == stock.libro.id)
                .one_or_none()
            )
            if modelo is None:
                raise ValueError("No se encuentra el stock para actualizar.")
            modelo.cantidad = stock.cantidad
            modelo.punto_reposicion = stock.punto_reposicion
            session.flush()
            cargado = (
                session.query(StockLibroModel)
                .options(*self._opciones())
                .filter(StockLibroModel.id == modelo.id)
                .one()
            )
            return a_stock(cargado)

    def eliminar(self, libro_id: int) -> bool:
        """Elimina el stock de un libro."""
        with self._db.transaccion() as session:
            modelo = (
                session.query(StockLibroModel)
                .filter(StockLibroModel.libro_id == libro_id)
                .one_or_none()
            )
            if modelo is None:
                return False
            session.delete(modelo)
            return True


class RepositorioCotizacionDolar(IRepositorioCotizacionDolar):
    """CRUD del histórico de cotizaciones del dólar."""

    def __init__(self, db: ConexionDB) -> None:
        self._db = db

    def _buscar(self, session, tipo_id: int, fecha: date) -> Optional[CotizacionDolarModel]:
        return (
            session.query(CotizacionDolarModel)
            .options(joinedload(CotizacionDolarModel.tipo_cotizacion))
            .filter(
                CotizacionDolarModel.tipo_cotizacion_id == tipo_id,
                CotizacionDolarModel.fecha == fecha,
            )
            .one_or_none()
        )

    def crear(self, cotizacion: CotizacionDolar) -> CotizacionDolar:
        """Inserta una cotización."""
        with self._db.transaccion() as session:
            if self._buscar(
                session, cotizacion.tipo_cotizacion.id, cotizacion.fecha
            ) is not None:
                raise ValueError("Ya existe una cotización para el mismo tipo y fecha.")
            modelo = CotizacionDolarModel(
                tipo_cotizacion_id=cotizacion.tipo_cotizacion.id,
                fecha=cotizacion.fecha,
                compra=cotizacion.compra,
                venta=cotizacion.venta,
            )
            session.add(modelo)
            session.flush()
            cargado = self._buscar(
                session, cotizacion.tipo_cotizacion.id, cotizacion.fecha
            )
            return a_cotizacion(cargado)

    def leer_por_tipo_y_fecha(
        self, tipo_id: int, fecha: date
    ) -> Optional[CotizacionDolar]:
        """Lee una cotización por tipo y fecha."""
        with self._db.transaccion() as session:
            modelo = self._buscar(session, tipo_id, fecha)
            if modelo is None:
                return None
            return a_cotizacion(modelo)

    def leer_historico_por_tipo(self, tipo_id: int) -> List[CotizacionDolar]:
        """Lee el histórico de un tipo, ordenado por fecha."""
        with self._db.transaccion() as session:
            filas = (
                session.query(CotizacionDolarModel)
                .options(joinedload(CotizacionDolarModel.tipo_cotizacion))
                .filter(CotizacionDolarModel.tipo_cotizacion_id == tipo_id)
                .order_by(CotizacionDolarModel.fecha)
                .all()
            )
            return [a_cotizacion(fila) for fila in filas]

    def leer_todas(self) -> List[CotizacionDolar]:
        """Lee todas las cotizaciones."""
        with self._db.transaccion() as session:
            filas = (
                session.query(CotizacionDolarModel)
                .options(joinedload(CotizacionDolarModel.tipo_cotizacion))
                .order_by(CotizacionDolarModel.fecha, CotizacionDolarModel.tipo_cotizacion_id)
                .all()
            )
            return [a_cotizacion(fila) for fila in filas]

    def actualizar(self, cotizacion: CotizacionDolar) -> CotizacionDolar:
        """Actualiza compra y venta de una cotización existente."""
        with self._db.transaccion() as session:
            modelo = self._buscar(
                session, cotizacion.tipo_cotizacion.id, cotizacion.fecha
            )
            if modelo is None:
                raise ValueError("No se encuentra la cotización para actualizar.")
            modelo.compra = cotizacion.compra
            modelo.venta = cotizacion.venta
            session.flush()
            return a_cotizacion(modelo)

    def eliminar(self, tipo_id: int, fecha: date) -> bool:
        """Elimina una cotización por tipo y fecha."""
        with self._db.transaccion() as session:
            modelo = self._buscar(session, tipo_id, fecha)
            if modelo is None:
                return False
            session.delete(modelo)
            return True


def a_extraccion(modelo: ExtraccionWebModel) -> ExtraccionWeb:
    """Convierte una fila de extracción web en la entidad."""
    return ExtraccionWeb(
        id=modelo.id,
        libro_id=modelo.libro_id,
        titulo_buscado=modelo.titulo_buscado,
        titulo_web=modelo.titulo_web,
        precio_web=modelo.precio_web,
        url_imagen=modelo.url_imagen or "",
        formas_pago=modelo.formas_pago or "",
        descripcion=modelo.descripcion or "",
        url_producto=modelo.url_producto,
        fecha_extraccion=modelo.fecha_extraccion,
        coincide=bool(modelo.coincide),
    )


def a_auditoria(modelo: AuditoriaModel) -> Auditoria:
    """Convierte una fila de auditoría en la entidad."""
    return Auditoria(
        id=modelo.id,
        accion=modelo.accion,
        fecha=modelo.fecha,
        detalles=modelo.detalles,
    )


class RepositorioExtraccionWeb:
    """CRUD de los resultados del scraper de Cúspide."""

    def __init__(self, db: ConexionDB) -> None:
        self._db = db

    def crear(self, entidad: ExtraccionWeb) -> ExtraccionWeb:
        """Inserta una extracción."""
        with self._db.transaccion() as session:
            modelo = ExtraccionWebModel(
                libro_id=entidad.libro_id,
                titulo_buscado=entidad.titulo_buscado,
                titulo_web=entidad.titulo_web,
                precio_web=entidad.precio_web,
                url_imagen=entidad.url_imagen,
                formas_pago=entidad.formas_pago[:500],
                descripcion=entidad.descripcion,
                url_producto=entidad.url_producto,
                fecha_extraccion=entidad.fecha_extraccion,
                coincide=1 if entidad.coincide else 0,
            )
            session.add(modelo)
            session.flush()
            return a_extraccion(modelo)

    def leer_por_id(self, id: int) -> Optional[ExtraccionWeb]:
        """Lee una extracción por id."""
        with self._db.transaccion() as session:
            modelo = session.get(ExtraccionWebModel, id)
            if modelo is None:
                return None
            return a_extraccion(modelo)

    def leer_todos(self) -> List[ExtraccionWeb]:
        """Lee todas las extracciones."""
        with self._db.transaccion() as session:
            filas = session.query(ExtraccionWebModel).order_by(ExtraccionWebModel.id).all()
            return [a_extraccion(fila) for fila in filas]

    def leer_por_libro(self, libro_id: int) -> List[ExtraccionWeb]:
        """Lee las extracciones de un libro del catálogo."""
        with self._db.transaccion() as session:
            filas = (
                session.query(ExtraccionWebModel)
                .filter(ExtraccionWebModel.libro_id == libro_id)
                .all()
            )
            return [a_extraccion(fila) for fila in filas]

    def actualizar(self, entidad: ExtraccionWeb) -> ExtraccionWeb:
        """Actualiza una extracción existente."""
        with self._db.transaccion() as session:
            modelo = session.get(ExtraccionWebModel, entidad.id)
            if modelo is None:
                raise ValueError("No se encuentra la extracción para actualizar.")
            modelo.titulo_web = entidad.titulo_web
            modelo.precio_web = entidad.precio_web
            modelo.url_imagen = entidad.url_imagen
            modelo.formas_pago = entidad.formas_pago[:500]
            modelo.descripcion = entidad.descripcion
            modelo.coincide = 1 if entidad.coincide else 0
            session.flush()
            return a_extraccion(modelo)

    def eliminar(self, id: int) -> bool:
        """Elimina una extracción por id."""
        with self._db.transaccion() as session:
            modelo = session.get(ExtraccionWebModel, id)
            if modelo is None:
                return False
            session.delete(modelo)
            return True

    def eliminar_todas(self) -> int:
        """Vacía la tabla antes de un scraping nuevo.

        Returns:
            int: Filas eliminadas.
        """
        with self._db.transaccion() as session:
            cantidad = session.query(ExtraccionWebModel).delete()
            return int(cantidad or 0)


class RepositorioAuditoria:
    """CRUD de la tabla de auditoría."""

    def __init__(self, db: ConexionDB) -> None:
        self._db = db

    def crear(self, entidad: Auditoria) -> Auditoria:
        """Inserta un registro de auditoría."""
        with self._db.transaccion() as session:
            modelo = AuditoriaModel(
                accion=entidad.accion[:80],
                fecha=entidad.fecha,
                detalles=entidad.detalles[:500],
            )
            if entidad.id:
                modelo.id = entidad.id
            session.add(modelo)
            session.flush()
            return a_auditoria(modelo)

    def leer_por_id(self, id: int) -> Optional[Auditoria]:
        """Lee un registro por id."""
        with self._db.transaccion() as session:
            modelo = session.get(AuditoriaModel, id)
            if modelo is None:
                return None
            return a_auditoria(modelo)

    def leer_todos(self) -> List[Auditoria]:
        """Lee el historial ordenado por fecha."""
        with self._db.transaccion() as session:
            filas = session.query(AuditoriaModel).order_by(AuditoriaModel.fecha, AuditoriaModel.id).all()
            return [a_auditoria(fila) for fila in filas]

    def actualizar(self, entidad: Auditoria) -> Auditoria:
        """Actualiza el detalle de un registro existente."""
        with self._db.transaccion() as session:
            modelo = session.get(AuditoriaModel, entidad.id)
            if modelo is None:
                raise ValueError("No se encuentra la auditoría para actualizar.")
            modelo.accion = entidad.accion[:80]
            modelo.fecha = entidad.fecha
            modelo.detalles = entidad.detalles[:500]
            session.flush()
            return a_auditoria(modelo)

    def eliminar(self, id: int) -> bool:
        """Elimina un registro de auditoría."""
        with self._db.transaccion() as session:
            modelo = session.get(AuditoriaModel, id)
            if modelo is None:
                return False
            session.delete(modelo)
            return True
''', encoding='utf-8')
print('Vista —', destino.name, destino.stat().st_size, 'bytes')
````

````python
#@title Ejercicio 05 — servicios de comparación, alertas y Excel
from pathlib import Path
destino = Path(RUTA_SRC) / 'book_manager/services/services.py'
destino.parent.mkdir(parents=True, exist_ok=True)
destino.write_text(r'''"""Lógica de negocio de Book Manager sobre la base relacional."""

import csv
import json
import os
import subprocess
import sys
from dataclasses import dataclass
from datetime import date
from datetime import datetime
from functools import wraps
from pathlib import Path
from types import FunctionType
from typing import List
from typing import Optional
from urllib.error import URLError
from urllib.request import Request
from urllib.request import urlopen

from dotenv import load_dotenv

from book_manager.database.connection import ConexionDB
from book_manager.entities.entities import Auditoria
from book_manager.entities.entities import CotizacionDolar
from book_manager.entities.entities import Editorial
from book_manager.entities.entities import Genero
from book_manager.entities.entities import Libro
from book_manager.entities.entities import Moneda
from book_manager.entities.entities import Precio
from book_manager.entities.entities import Stock
from book_manager.entities.entities import TipoCotizacion
from book_manager.repositories.repositories import RepositorioAuditoria
from book_manager.repositories.repositories import RepositorioCotizacionDolar
from book_manager.repositories.repositories import RepositorioEditorial
from book_manager.repositories.repositories import RepositorioExtraccionWeb
from book_manager.repositories.repositories import RepositorioGenero
from book_manager.repositories.repositories import RepositorioLibro
from book_manager.repositories.repositories import RepositorioMoneda
from book_manager.repositories.repositories import RepositorioPrecio
from book_manager.repositories.repositories import RepositorioStock
from book_manager.repositories.repositories import RepositorioTipoCotizacion

MAPA_CASA_DOLAR = {
    "oficial": "Oficial",
    "blue": "Blue",
    "bolsa": "MEP",
    "contadoconliqui": "CCL",
    "mayorista": "Mayorista",
    "tarjeta": "Tarjeta",
    "cripto": "Cripto",
}


def _mismo_texto(izquierdo: str, derecho: str) -> bool:
    """Compara dos textos sin distinguir mayúsculas."""
    return izquierdo.strip().casefold() == derecho.strip().casefold()


def cargar_env() -> None:
    """Carga el archivo .env de la raíz del proyecto book_manager."""
    ruta = Path(__file__).resolve().parents[3] / ".env"
    load_dotenv(ruta)


class GeneroServicio:
    """Reglas de alta, lectura, modificación y borrado de géneros."""

    def __init__(self, repositorio: RepositorioGenero, libros: RepositorioLibro) -> None:
        self.__repositorio = repositorio
        self.__libros = libros

    def crear(self, entidad: Genero) -> Genero:
        """Crea un género si el nombre no está repetido."""
        if self.buscar_por_nombre(entidad.nombre) is not None:
            raise ValueError(f"Ya existe un género llamado {entidad.nombre}.")
        return self.__repositorio.crear(entidad)

    def leer_por_id(self, id: int) -> Optional[Genero]:
        """Lee un género por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[Genero]:
        """Lee todos los géneros."""
        return self.__repositorio.leer_todos()

    def buscar_por_nombre(self, nombre: str) -> Optional[Genero]:
        """Busca un género por nombre."""
        for genero in self.__repositorio.leer_todos():
            if _mismo_texto(genero.nombre, nombre):
                return genero
        return None

    def actualizar(self, entidad: Genero) -> Genero:
        """Actualiza un género y evita nombres duplicados."""
        if self.__repositorio.leer_por_id(entidad.id) is None:
            raise ValueError("No se encuentra el género para actualizar.")
        otro = self.buscar_por_nombre(entidad.nombre)
        if otro is not None and otro.id != entidad.id:
            raise ValueError(f"Ya existe un género llamado {entidad.nombre}.")
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina un género que no tenga libros asociados."""
        if any(libro.genero.id == id for libro in self.__libros.leer_todos()):
            raise ValueError(
                "No se puede eliminar el género porque hay libros asociados."
            )
        return self.__repositorio.eliminar(id)


class EditorialServicio:
    """Reglas de alta, lectura, modificación y borrado de editoriales."""

    def __init__(
        self, repositorio: RepositorioEditorial, libros: RepositorioLibro
    ) -> None:
        self.__repositorio = repositorio
        self.__libros = libros

    def crear(self, entidad: Editorial) -> Editorial:
        """Crea una editorial si el nombre no está repetido."""
        if self.buscar_por_nombre(entidad.nombre) is not None:
            raise ValueError(f"Ya existe la editorial {entidad.nombre}.")
        return self.__repositorio.crear(entidad)

    def leer_por_id(self, id: int) -> Optional[Editorial]:
        """Lee una editorial por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[Editorial]:
        """Lee todas las editoriales."""
        return self.__repositorio.leer_todos()

    def buscar_por_nombre(self, nombre: str) -> Optional[Editorial]:
        """Busca una editorial por nombre."""
        for editorial in self.__repositorio.leer_todos():
            if _mismo_texto(editorial.nombre, nombre):
                return editorial
        return None

    def actualizar(self, entidad: Editorial) -> Editorial:
        """Actualiza una editorial."""
        if self.__repositorio.leer_por_id(entidad.id) is None:
            raise ValueError("No se encuentra la editorial para actualizar.")
        otro = self.buscar_por_nombre(entidad.nombre)
        if otro is not None and otro.id != entidad.id:
            raise ValueError(f"Ya existe la editorial {entidad.nombre}.")
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina una editorial que no tenga libros asociados."""
        if any(libro.editorial.id == id for libro in self.__libros.leer_todos()):
            raise ValueError(
                "No se puede eliminar la editorial porque hay libros asociados."
            )
        return self.__repositorio.eliminar(id)


class MonedaServicio:
    """Reglas de alta, lectura, modificación y borrado de monedas."""

    def __init__(self, repositorio: RepositorioMoneda, precios: RepositorioPrecio) -> None:
        self.__repositorio = repositorio
        self.__precios = precios

    def crear(self, entidad: Moneda) -> Moneda:
        """Crea una moneda si el código no está repetido."""
        if self.buscar_por_codigo(entidad.codigo) is not None:
            raise ValueError(f"Ya existe la moneda {entidad.codigo}.")
        return self.__repositorio.crear(entidad)

    def leer_por_id(self, id: int) -> Optional[Moneda]:
        """Lee una moneda por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[Moneda]:
        """Lee todas las monedas."""
        return self.__repositorio.leer_todos()

    def buscar_por_codigo(self, codigo: str) -> Optional[Moneda]:
        """Busca una moneda por su código."""
        buscado = codigo.strip().upper()
        for moneda in self.__repositorio.leer_todos():
            if moneda.codigo == buscado:
                return moneda
        return None

    def actualizar(self, entidad: Moneda) -> Moneda:
        """Actualiza una moneda."""
        if self.__repositorio.leer_por_id(entidad.id) is None:
            raise ValueError("No se encuentra la moneda para actualizar.")
        otro = self.buscar_por_codigo(entidad.codigo)
        if otro is not None and otro.id != entidad.id:
            raise ValueError(f"Ya existe la moneda {entidad.codigo}.")
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina una moneda que no esté usada en precios."""
        if any(precio.moneda.id == id for precio in self.__precios.leer_todos()):
            raise ValueError(
                "No se puede eliminar la moneda porque hay precios asociados."
            )
        return self.__repositorio.eliminar(id)


class TipoCotizacionServicio:
    """Reglas de alta, lectura, modificación y borrado de tipos de cotización."""

    def __init__(
        self,
        repositorio: RepositorioTipoCotizacion,
        cotizaciones: RepositorioCotizacionDolar,
    ) -> None:
        self.__repositorio = repositorio
        self.__cotizaciones = cotizaciones

    def crear(self, entidad: TipoCotizacion) -> TipoCotizacion:
        """Crea un tipo de cotización si el nombre no está repetido."""
        if self.buscar_por_nombre(entidad.nombre) is not None:
            raise ValueError(f"Ya existe el tipo de cotización {entidad.nombre}.")
        return self.__repositorio.crear(entidad)

    def leer_por_id(self, id: int) -> Optional[TipoCotizacion]:
        """Lee un tipo de cotización por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[TipoCotizacion]:
        """Lee todos los tipos de cotización."""
        return self.__repositorio.leer_todos()

    def buscar_por_nombre(self, nombre: str) -> Optional[TipoCotizacion]:
        """Busca un tipo de cotización por nombre."""
        for tipo in self.__repositorio.leer_todos():
            if _mismo_texto(tipo.nombre, nombre):
                return tipo
        return None

    def actualizar(self, entidad: TipoCotizacion) -> TipoCotizacion:
        """Actualiza un tipo de cotización."""
        if self.__repositorio.leer_por_id(entidad.id) is None:
            raise ValueError("No se encuentra el tipo de cotización para actualizar.")
        otro = self.buscar_por_nombre(entidad.nombre)
        if otro is not None and otro.id != entidad.id:
            raise ValueError(f"Ya existe el tipo de cotización {entidad.nombre}.")
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina un tipo que no tenga cotizaciones cargadas."""
        if self.__cotizaciones.leer_historico_por_tipo(id):
            raise ValueError(
                "No se puede eliminar el tipo porque tiene cotizaciones cargadas."
            )
        return self.__repositorio.eliminar(id)


class LibroServicio:
    """Reglas de alta, lectura, modificación y borrado de libros."""

    def __init__(
        self,
        repositorio: RepositorioLibro,
        generos: RepositorioGenero,
        editoriales: RepositorioEditorial,
        precios: RepositorioPrecio,
        stocks: RepositorioStock,
    ) -> None:
        self.__repositorio = repositorio
        self.__generos = generos
        self.__editoriales = editoriales
        self.__precios = precios
        self.__stocks = stocks

    def crear(self, entidad: Libro) -> Libro:
        """Crea un libro si el ISBN es único y las relaciones existen."""
        self.__validar_relaciones(entidad)
        if self.buscar_por_isbn(entidad.isbn) is not None:
            raise ValueError(f"Ya existe un libro con ISBN {entidad.isbn}.")
        return self.__repositorio.crear(entidad)

    def leer_por_id(self, id: int) -> Optional[Libro]:
        """Lee un libro por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[Libro]:
        """Lee todos los libros."""
        return self.__repositorio.leer_todos()

    def buscar_por_isbn(self, isbn: str) -> Optional[Libro]:
        """Busca un libro por ISBN."""
        for libro in self.__repositorio.leer_todos():
            if _mismo_texto(libro.isbn, isbn):
                return libro
        return None

    def actualizar(self, entidad: Libro) -> Libro:
        """Actualiza un libro."""
        if self.__repositorio.leer_por_id(entidad.id) is None:
            raise ValueError("No se encuentra el libro para actualizar.")
        self.__validar_relaciones(entidad)
        otro = self.buscar_por_isbn(entidad.isbn)
        if otro is not None and otro.id != entidad.id:
            raise ValueError(f"Ya existe un libro con ISBN {entidad.isbn}.")
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina un libro que no tenga precio ni stock."""
        if any(precio.libro.id == id for precio in self.__precios.leer_todos()):
            raise ValueError(
                "No se puede eliminar el libro porque tiene precios asociados."
            )
        if self.__stocks.leer_por_libro(id) is not None:
            raise ValueError(
                "No se puede eliminar el libro porque tiene stock asociado."
            )
        return self.__repositorio.eliminar(id)

    def __validar_relaciones(self, entidad: Libro) -> None:
        if self.__generos.leer_por_id(entidad.genero.id) is None:
            raise ValueError("El género indicado no existe.")
        if self.__editoriales.leer_por_id(entidad.editorial.id) is None:
            raise ValueError("La editorial indicada no existe.")


class PrecioServicio:
    """Reglas de alta, lectura, modificación y borrado de precios."""

    def __init__(
        self,
        repositorio: RepositorioPrecio,
        libros: RepositorioLibro,
        monedas: RepositorioMoneda,
    ) -> None:
        self.__repositorio = repositorio
        self.__libros = libros
        self.__monedas = monedas

    def crear(self, entidad: Precio) -> Precio:
        """Crea un precio si el libro y la moneda existen y no hay duplicado."""
        self.__validar_relaciones(entidad)
        if self.buscar(entidad.libro.id, entidad.moneda.id) is not None:
            raise ValueError("Ya existe un precio para ese libro y esa moneda.")
        return self.__repositorio.crear(entidad)

    def leer_por_id(self, id: int) -> Optional[Precio]:
        """Lee un precio por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[Precio]:
        """Lee todos los precios."""
        return self.__repositorio.leer_todos()

    def buscar(self, libro_id: int, moneda_id: int) -> Optional[Precio]:
        """Busca el precio de un libro en una moneda."""
        for precio in self.__repositorio.leer_todos():
            if precio.libro.id == libro_id and precio.moneda.id == moneda_id:
                return precio
        return None

    def actualizar(self, entidad: Precio) -> Precio:
        """Actualiza un precio."""
        if self.__repositorio.leer_por_id(entidad.id) is None:
            raise ValueError("No se encuentra el precio para actualizar.")
        self.__validar_relaciones(entidad)
        otro = self.buscar(entidad.libro.id, entidad.moneda.id)
        if otro is not None and otro.id != entidad.id:
            raise ValueError("Ya existe un precio para ese libro y esa moneda.")
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina un precio por id."""
        return self.__repositorio.eliminar(id)

    def __validar_relaciones(self, entidad: Precio) -> None:
        if self.__libros.leer_por_id(entidad.libro.id) is None:
            raise ValueError("El libro indicado no existe.")
        if self.__monedas.leer_por_id(entidad.moneda.id) is None:
            raise ValueError("La moneda indicada no existe.")


class StockServicio:
    """Reglas de alta, lectura, modificación y borrado de stock."""

    def __init__(self, repositorio: RepositorioStock, libros: RepositorioLibro) -> None:
        self.__repositorio = repositorio
        self.__libros = libros

    def crear(self, stock: Stock) -> Stock:
        """Crea el stock de un libro."""
        self.__validar_libro(stock.libro.id)
        if self.__repositorio.leer_por_libro(stock.libro.id) is not None:
            raise ValueError("Ya existe stock para ese libro.")
        return self.__repositorio.crear(stock)

    def leer_por_libro(self, libro_id: int) -> Optional[Stock]:
        """Lee el stock de un libro."""
        return self.__repositorio.leer_por_libro(libro_id)

    def leer_todos(self) -> List[Stock]:
        """Lee todos los stocks."""
        return self.__repositorio.leer_todos()

    def actualizar(self, stock: Stock) -> Stock:
        """Actualiza el stock de un libro."""
        self.__validar_libro(stock.libro.id)
        return self.__repositorio.actualizar(stock)

    def eliminar(self, libro_id: int) -> bool:
        """Elimina el stock de un libro."""
        return self.__repositorio.eliminar(libro_id)

    def __validar_libro(self, libro_id: int) -> None:
        if self.__libros.leer_por_id(libro_id) is None:
            raise ValueError("El libro indicado no existe.")


class ServicioCotizacionDolar:
    """Reglas de cotización y descarga desde la API del dólar."""

    def __init__(
        self,
        repositorio: RepositorioCotizacionDolar,
        tipos: RepositorioTipoCotizacion,
    ) -> None:
        self.__repositorio = repositorio
        self.__tipos = tipos

    def crear(self, cotizacion: CotizacionDolar) -> CotizacionDolar:
        """Crea una cotización si el tipo existe y la fecha no está repetida."""
        self.__validar_tipo(cotizacion.tipo_cotizacion.id)
        existente = self.__repositorio.leer_por_tipo_y_fecha(
            cotizacion.tipo_cotizacion.id,
            cotizacion.fecha,
        )
        if existente is not None:
            raise ValueError("Ya existe una cotización para ese tipo y esa fecha.")
        return self.__repositorio.crear(cotizacion)

    def leer_por_tipo_y_fecha(
        self, tipo_id: int, fecha: date
    ) -> Optional[CotizacionDolar]:
        """Lee una cotización por tipo y fecha."""
        return self.__repositorio.leer_por_tipo_y_fecha(tipo_id, fecha)

    def leer_historico_por_tipo(self, tipo_id: int) -> List[CotizacionDolar]:
        """Lee el histórico de un tipo de cotización."""
        self.__validar_tipo(tipo_id)
        return self.__repositorio.leer_historico_por_tipo(tipo_id)

    def leer_todas(self) -> List[CotizacionDolar]:
        """Lee todas las cotizaciones."""
        return self.__repositorio.leer_todas()

    def actualizar(self, cotizacion: CotizacionDolar) -> CotizacionDolar:
        """Actualiza compra y venta de una cotización existente."""
        self.__validar_tipo(cotizacion.tipo_cotizacion.id)
        return self.__repositorio.actualizar(cotizacion)

    def eliminar(self, tipo_id: int, fecha: date) -> bool:
        """Elimina una cotización por tipo y fecha."""
        return self.__repositorio.eliminar(tipo_id, fecha)

    def obtener_cotizaciones(self) -> List[CotizacionDolar]:
        """Descarga las cotizaciones de la API y las registra en la base.

        La URL se lee de la variable API_URL del archivo .env.

        Returns:
            List[CotizacionDolar]: Cotizaciones creadas o actualizadas.

        Raises:
            ValueError: Si falta API_URL, la respuesta no es una lista
                o no hay conexión con la API (URLError).
        """
        cargar_env()
        url = os.getenv("API_URL", "").strip()
        if url == "":
            raise ValueError("Falta API_URL en el archivo .env.")
        pedido = Request(
            url,
            headers={"User-Agent": "BookManager/1.0 (Grupo 28; educational)"},
        )
        try:
            with urlopen(pedido, timeout=30) as respuesta:
                payload = json.loads(respuesta.read().decode("utf-8"))
        except URLError as error:
            raise ValueError(
                "No se pudo conectar con la API del dólar. "
                f"Revisá la red o API_URL. Detalle: {error.reason}"
            ) from error
        if not isinstance(payload, list):
            raise ValueError("La API del dólar no devolvió una lista.")
        registradas: List[CotizacionDolar] = []
        for item in payload:
            casa = str(item.get("casa", "")).strip().lower()
            nombre = MAPA_CASA_DOLAR.get(casa)
            if nombre is None:
                continue
            compra = item.get("compra")
            venta = item.get("venta")
            if compra is None or venta is None:
                continue
            if float(compra) <= 0 or float(venta) <= 0 or float(compra) > float(venta):
                continue
            fecha = _fecha_de_api(item.get("fechaActualizacion"))
            tipo = self.__tipo_para(nombre, str(item.get("nombre") or nombre))
            cotizacion = CotizacionDolar(
                tipo_cotizacion=tipo,
                fecha=fecha,
                compra=float(compra),
                venta=float(venta),
            )
            existente = self.__repositorio.leer_por_tipo_y_fecha(tipo.id, fecha)
            if existente is None:
                registradas.append(self.__repositorio.crear(cotizacion))
            else:
                registradas.append(self.__repositorio.actualizar(cotizacion))
        return registradas

    def __tipo_para(self, nombre: str, descripcion: str) -> TipoCotizacion:
        existente = None
        for tipo in self.__tipos.leer_todos():
            if _mismo_texto(tipo.nombre, nombre):
                existente = tipo
                break
        if existente is not None:
            return existente
        return self.__tipos.crear(
            TipoCotizacion(id=0, nombre=nombre, descripcion=descripcion)
        )

    def __validar_tipo(self, tipo_id: int) -> None:
        if self.__tipos.leer_por_id(tipo_id) is None:
            raise ValueError("El tipo de cotización indicado no existe.")


def _fecha_de_api(valor: Optional[str]) -> date:
    """Toma la fecha de la marca de tiempo que devuelve la API."""
    if not valor:
        return date.today()
    texto = str(valor).replace("Z", "")
    try:
        return datetime.fromisoformat(texto).date()
    except ValueError:
        return date.fromisoformat(texto[:10])


@dataclass(frozen=True)
class ComparacionCuspide:
    """Comparación entre el precio público de Cúspide y el costo de reposición."""

    titulo: str
    autor: str
    precio_cuspide_ars: float
    costo_usd: float
    tipo_cotizacion: str
    cotizacion_venta: float
    costo_reposicion_ars: float

    @property
    def diferencia_ars(self) -> float:
        """Diferencia entre el precio de Cúspide y el costo en pesos."""
        return round(self.precio_cuspide_ars - self.costo_reposicion_ars, 2)

    @property
    def cubre_el_costo(self) -> bool:
        """Indica si el precio de Cúspide alcanza para cubrir la reposición."""
        return self.diferencia_ars >= 0


@dataclass(frozen=True)
class LineaStock:
    """Fila del reporte de reposición."""

    titulo: str
    autor: str
    cantidad: int
    punto_reposicion: int


@dataclass(frozen=True)
class LineaHistorico:
    """Fila del histórico de una cotización."""

    fecha: date
    tipo: str
    compra: float
    venta: float


@dataclass(frozen=True)
class LineaBimonetaria:
    """Precio de un libro en pesos y en otra moneda."""

    titulo: str
    autor: str
    precio_ars: Optional[float]
    codigo_moneda: str
    precio_otra: Optional[float]


class ReporteServicio:
    """Reportes de cotización, Cúspide, stock, histórico y precios bimonetarios."""

    def __init__(
        self,
        libros: RepositorioLibro,
        precios: RepositorioPrecio,
        monedas: RepositorioMoneda,
        stocks: RepositorioStock,
        cotizaciones: RepositorioCotizacionDolar,
        tipos: RepositorioTipoCotizacion,
    ) -> None:
        self.__libros = libros
        self.__precios = precios
        self.__monedas = monedas
        self.__stocks = stocks
        self.__cotizaciones = cotizaciones
        self.__tipos = tipos

    def comparar_con_cuspide(self, tipo_id: int) -> List[ComparacionCuspide]:
        """Cotiza el costo en dólares y lo compara con el precio de Cúspide."""
        tipo = self.__tipos.leer_por_id(tipo_id)
        if tipo is None:
            raise ValueError("El tipo de cotización indicado no existe.")
        historico = self.__cotizaciones.leer_historico_por_tipo(tipo_id)
        if not historico:
            raise ValueError("No hay cotizaciones para el tipo indicado.")
        cotizacion = historico[-1]
        moneda_ars = self.__moneda_por_codigo("ARS")
        moneda_usd = self.__moneda_por_codigo("USD")
        filas: List[ComparacionCuspide] = []
        for libro in self.__libros.leer_todos():
            precio_ars = self.__importe(libro.id, moneda_ars.id)
            costo_usd = self.__importe(libro.id, moneda_usd.id)
            if precio_ars is None or costo_usd is None:
                continue
            costo_ars = round(costo_usd * cotizacion.venta, 2)
            filas.append(
                ComparacionCuspide(
                    titulo=libro.titulo,
                    autor=libro.autor,
                    precio_cuspide_ars=precio_ars,
                    costo_usd=costo_usd,
                    tipo_cotizacion=tipo.nombre,
                    cotizacion_venta=cotizacion.venta,
                    costo_reposicion_ars=costo_ars,
                )
            )
        return filas

    def stock_a_reponer(self) -> List[LineaStock]:
        """Devuelve los libros con stock en el punto de reposición o por debajo."""
        filas: List[LineaStock] = []
        for stock in self.__stocks.leer_todos():
            if stock.cantidad > stock.punto_reposicion:
                continue
            filas.append(
                LineaStock(
                    titulo=stock.libro.titulo,
                    autor=stock.libro.autor,
                    cantidad=stock.cantidad,
                    punto_reposicion=stock.punto_reposicion,
                )
            )
        return filas

    def historico(self, tipo_id: int) -> List[LineaHistorico]:
        """Devuelve el histórico de un tipo de dólar, ordenado por fecha."""
        tipo = self.__tipos.leer_por_id(tipo_id)
        if tipo is None:
            raise ValueError("El tipo de cotización indicado no existe.")
        return [
            LineaHistorico(
                fecha=item.fecha,
                tipo=tipo.nombre,
                compra=item.compra,
                venta=item.venta,
            )
            for item in self.__cotizaciones.leer_historico_por_tipo(tipo_id)
        ]

    def lista_bimonetaria(self, codigo_moneda: str) -> List[LineaBimonetaria]:
        """Lista cada libro en pesos y en la moneda elegida.

        Args:
            codigo_moneda (str): Código de tres letras de la segunda moneda.

        Returns:
            List[LineaBimonetaria]: Una fila por libro.
        """
        moneda_ars = self.__moneda_por_codigo("ARS")
        otra = self.__moneda_por_codigo(codigo_moneda)
        filas: List[LineaBimonetaria] = []
        for libro in self.__libros.leer_todos():
            filas.append(
                LineaBimonetaria(
                    titulo=libro.titulo,
                    autor=libro.autor,
                    precio_ars=self.__importe(libro.id, moneda_ars.id),
                    codigo_moneda=otra.codigo,
                    precio_otra=self.__importe(libro.id, otra.id),
                )
            )
        return filas

    def exportar_precios_csv(self, ruta: Optional[Path] = None) -> Path:
        """Exporta el precio de cada libro en todas las monedas cargadas.

        Args:
            ruta (Optional[Path]): Destino del CSV. Si no se indica, se usa
                book_manager/export/precios.csv.

        Returns:
            Path: Ruta del archivo generado.
        """
        if ruta is None:
            ruta = Path(__file__).resolve().parents[3] / "export" / "precios.csv"
        ruta.parent.mkdir(parents=True, exist_ok=True)
        with ruta.open("w", encoding="utf-8", newline="") as archivo:
            escritor = csv.DictWriter(
                archivo,
                fieldnames=["isbn", "titulo", "autor", "moneda", "importe"],
            )
            escritor.writeheader()
            for precio in self.__precios.leer_todos():
                escritor.writerow(
                    {
                        "isbn": precio.libro.isbn,
                        "titulo": precio.libro.titulo,
                        "autor": precio.libro.autor,
                        "moneda": precio.moneda.codigo,
                        "importe": f"{precio.importe:.2f}",
                    }
                )
        return ruta

    def __moneda_por_codigo(self, codigo: str) -> Moneda:
        buscado = codigo.strip().upper()
        for moneda in self.__monedas.leer_todos():
            if moneda.codigo == buscado:
                return moneda
        raise ValueError(f"No está cargada la moneda {codigo}.")

    def __importe(self, libro_id: int, moneda_id: int) -> Optional[float]:
        for precio in self.__precios.leer_todos():
            if precio.libro.id == libro_id and precio.moneda.id == moneda_id:
                return precio.importe
        return None


@dataclass(frozen=True)
class LineaPrecioWeb:
    """Comparación entre el precio interno y el precio publicado en Cúspide."""

    libro: str
    precio_interno: Optional[float]
    precio_web: Optional[float]
    diferencia: Optional[float]
    fecha_extraccion: Optional[datetime]


def armar_comparaciones(
    libros: RepositorioLibro,
    precios: RepositorioPrecio,
    extracciones: RepositorioExtraccionWeb,
) -> List[LineaPrecioWeb]:
    """Arma una fila por libro del catálogo.

    Usa la extracción cuyo título coincide con el del catálogo.
    """
    por_libro: dict[int, List] = {}
    for fila in extracciones.leer_todos():
        por_libro.setdefault(fila.libro_id, []).append(fila)
    resultado = []
    for libro in libros.leer_todos():
        interno = _importe_de(precios, libro.id, "ARS")
        candidatas = por_libro.get(libro.id, [])
        elegida = _elegir_extraccion(candidatas, libro.titulo, interno)
        web = elegida.precio_web if elegida is not None else None
        fecha = elegida.fecha_extraccion if elegida is not None else None
        diferencia = None
        if interno is not None and web is not None:
            diferencia = round(web - interno, 2)
        resultado.append(
            LineaPrecioWeb(
                libro=f"{libro.titulo} ({libro.autor})",
                precio_interno=interno,
                precio_web=web,
                diferencia=diferencia,
                fecha_extraccion=fecha,
            )
        )
    return resultado


def _elegir_extraccion(candidatas, titulo: str, interno: Optional[float]):
    """Prefiere el título exacto y, si hay varios, el precio más cercano."""
    buscado = " ".join(titulo.split()).casefold()
    exactas = [
        fila
        for fila in candidatas
        if " ".join(fila.titulo_web.split()).casefold() == buscado
    ]
    pool = exactas or [fila for fila in candidatas if fila.coincide]
    if not pool:
        return None

    def distancia(fila) -> float:
        if fila.precio_web is None or interno is None:
            return 10**12
        return abs(fila.precio_web - interno)

    return min(pool, key=distancia)


def _importe_de(
    precios: RepositorioPrecio,
    libro_id: int,
    codigo: str,
) -> Optional[float]:
    for precio in precios.leer_todos():
        if precio.libro.id == libro_id and precio.moneda.codigo == codigo:
            return precio.importe
    return None


class ServicioScraping:
    """Ejecuta el spider y deja el CSV de alertas."""

    def __init__(
        self,
        extracciones: RepositorioExtraccionWeb,
        libros: RepositorioLibro,
        precios: RepositorioPrecio,
    ) -> None:
        self.__extracciones = extracciones
        self.__libros = libros
        self.__precios = precios

    def ejecutar(self, umbral: float, limite: int = 0) -> Path:
        """Pide el umbral, corre el scraper y escribe las alertas.

        Args:
            umbral (float): Diferencia mínima, en pesos, para alertar.
            limite (int): Títulos a buscar. Cero recorre el catálogo.

        Returns:
            Path: CSV de alertas.

        Raises:
            ValueError: Si el umbral es negativo o el scraper falla.
        """
        if umbral < 0:
            raise ValueError("La diferencia tiene que ser mayor o igual a cero.")
        self._correr(limite)
        return self.generar_alertas_csv(umbral)

    def generar_alertas_csv(self, umbral: float) -> Path:
        """Escribe el CSV descargable con los libros que superan el umbral.

        Args:
            umbral (float): Diferencia mínima en pesos.

        Returns:
            Path: Ruta del CSV.
        """
        if umbral < 0:
            raise ValueError("La diferencia tiene que ser mayor o igual a cero.")
        carpeta = Path(__file__).resolve().parents[1] / "reportes"
        carpeta.mkdir(parents=True, exist_ok=True)
        destino = carpeta / "alertas.csv"
        filas = [
            fila
            for fila in armar_comparaciones(
                self.__libros, self.__precios, self.__extracciones
            )
            if fila.diferencia is not None and abs(fila.diferencia) >= umbral
        ]
        with destino.open("w", encoding="utf-8", newline="") as archivo:
            escritor = csv.DictWriter(
                archivo,
                fieldnames=[
                    "libro",
                    "precio_interno",
                    "precio_web",
                    "diferencia",
                    "umbral",
                    "fecha_extraccion",
                ],
            )
            escritor.writeheader()
            for fila in filas:
                fecha = (
                    fila.fecha_extraccion.strftime("%Y-%m-%d %H:%M:%S")
                    if fila.fecha_extraccion is not None
                    else ""
                )
                escritor.writerow(
                    {
                        "libro": fila.libro,
                        "precio_interno": fila.precio_interno,
                        "precio_web": fila.precio_web,
                        "diferencia": fila.diferencia,
                        "umbral": umbral,
                        "fecha_extraccion": fecha,
                    }
                )
        return destino

    def generar_reporte_excel(self) -> Path:
        """Genera el Excel de precios internos y web.

        Returns:
            Path: Ruta del archivo .xlsx.
        """
        from openpyxl import Workbook

        carpeta = Path(__file__).resolve().parents[1] / "reportes"
        carpeta.mkdir(parents=True, exist_ok=True)
        destino = carpeta / "comparacion_precios.xlsx"
        libro = Workbook()
        hoja = libro.active
        hoja.title = "Precios"
        hoja.append(
            [
                "Libro",
                "Precio interno",
                "Precio web",
                "Diferencia",
                "Fecha de extracción",
            ]
        )
        for fila in armar_comparaciones(
            self.__libros, self.__precios, self.__extracciones
        ):
            fecha = (
                fila.fecha_extraccion.strftime("%Y-%m-%d %H:%M:%S")
                if fila.fecha_extraccion is not None
                else ""
            )
            hoja.append(
                [
                    fila.libro,
                    fila.precio_interno,
                    fila.precio_web,
                    fila.diferencia,
                    fecha,
                ]
            )
        libro.save(destino)
        return destino

    def _correr(self, limite: int) -> None:
        entorno = os.environ.copy()
        entorno["PYTHONPATH"] = str(Path(__file__).resolve().parents[2])
        resultado = subprocess.run(
            [sys.executable, "-m", "book_manager.scraper.ejecutar", str(limite)],
            env=entorno,
            cwd=str(Path(__file__).resolve().parents[4]),
        )
        if resultado.returncode != 0:
            raise ValueError("El scraper de Cúspide no pudo completar la búsqueda.")


class ServicioAuditoria:
    """Lectura y alta del historial de auditoría."""

    def __init__(self, repositorio: RepositorioAuditoria) -> None:
        self.__repositorio = repositorio

    def crear(self, accion: str, detalles: str) -> Auditoria:
        """Registra una acción."""
        return self.__repositorio.crear(
            Auditoria(
                id=0,
                accion=accion,
                fecha=datetime.now(),
                detalles=detalles,
            )
        )

    def leer_por_id(self, id: int) -> Optional[Auditoria]:
        """Lee un registro por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[Auditoria]:
        """Devuelve el historial completo."""
        return self.__repositorio.leer_todos()

    def actualizar(self, entidad: Auditoria) -> Auditoria:
        """Actualiza un registro."""
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina un registro."""
        return self.__repositorio.eliminar(id)


class Contexto:
    """Servicios que usa la consola y la demostración."""

    generos: GeneroServicio
    editoriales: EditorialServicio
    monedas: MonedaServicio
    tipos_cotizacion: TipoCotizacionServicio
    libros: LibroServicio
    precios: PrecioServicio
    stocks: StockServicio
    cotizaciones: ServicioCotizacionDolar
    reportes: ReporteServicio
    scraping: ServicioScraping
    auditoria: ServicioAuditoria


ServicioGenero = GeneroServicio
ServicioEditorial = EditorialServicio
ServicioMoneda = MonedaServicio
ServicioTipoCotizacion = TipoCotizacionServicio
ServicioLibro = LibroServicio
ServicioStock = StockServicio
''', encoding='utf-8')
print('Vista —', destino.name, destino.stat().st_size, 'bytes')
````

````python
#@title Ejercicio 06 — decorador de auditoría
from pathlib import Path
destino = Path(RUTA_SRC) / 'book_manager/services/services.py'
destino.parent.mkdir(parents=True, exist_ok=True)
destino.write_text(r'''"""Lógica de negocio de Book Manager sobre la base relacional."""

import csv
import json
import os
import subprocess
import sys
from dataclasses import dataclass
from datetime import date
from datetime import datetime
from functools import wraps
from pathlib import Path
from types import FunctionType
from typing import List
from typing import Optional
from urllib.error import URLError
from urllib.request import Request
from urllib.request import urlopen

from dotenv import load_dotenv

from book_manager.database.connection import ConexionDB
from book_manager.entities.entities import Auditoria
from book_manager.entities.entities import CotizacionDolar
from book_manager.entities.entities import Editorial
from book_manager.entities.entities import Genero
from book_manager.entities.entities import Libro
from book_manager.entities.entities import Moneda
from book_manager.entities.entities import Precio
from book_manager.entities.entities import Stock
from book_manager.entities.entities import TipoCotizacion
from book_manager.repositories.repositories import RepositorioAuditoria
from book_manager.repositories.repositories import RepositorioCotizacionDolar
from book_manager.repositories.repositories import RepositorioEditorial
from book_manager.repositories.repositories import RepositorioExtraccionWeb
from book_manager.repositories.repositories import RepositorioGenero
from book_manager.repositories.repositories import RepositorioLibro
from book_manager.repositories.repositories import RepositorioMoneda
from book_manager.repositories.repositories import RepositorioPrecio
from book_manager.repositories.repositories import RepositorioStock
from book_manager.repositories.repositories import RepositorioTipoCotizacion

MAPA_CASA_DOLAR = {
    "oficial": "Oficial",
    "blue": "Blue",
    "bolsa": "MEP",
    "contadoconliqui": "CCL",
    "mayorista": "Mayorista",
    "tarjeta": "Tarjeta",
    "cripto": "Cripto",
}


def _mismo_texto(izquierdo: str, derecho: str) -> bool:
    """Compara dos textos sin distinguir mayúsculas."""
    return izquierdo.strip().casefold() == derecho.strip().casefold()


def cargar_env() -> None:
    """Carga el archivo .env de la raíz del proyecto book_manager."""
    ruta = Path(__file__).resolve().parents[3] / ".env"
    load_dotenv(ruta)


class GeneroServicio:
    """Reglas de alta, lectura, modificación y borrado de géneros."""

    def __init__(self, repositorio: RepositorioGenero, libros: RepositorioLibro) -> None:
        self.__repositorio = repositorio
        self.__libros = libros

    def crear(self, entidad: Genero) -> Genero:
        """Crea un género si el nombre no está repetido."""
        if self.buscar_por_nombre(entidad.nombre) is not None:
            raise ValueError(f"Ya existe un género llamado {entidad.nombre}.")
        return self.__repositorio.crear(entidad)

    def leer_por_id(self, id: int) -> Optional[Genero]:
        """Lee un género por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[Genero]:
        """Lee todos los géneros."""
        return self.__repositorio.leer_todos()

    def buscar_por_nombre(self, nombre: str) -> Optional[Genero]:
        """Busca un género por nombre."""
        for genero in self.__repositorio.leer_todos():
            if _mismo_texto(genero.nombre, nombre):
                return genero
        return None

    def actualizar(self, entidad: Genero) -> Genero:
        """Actualiza un género y evita nombres duplicados."""
        if self.__repositorio.leer_por_id(entidad.id) is None:
            raise ValueError("No se encuentra el género para actualizar.")
        otro = self.buscar_por_nombre(entidad.nombre)
        if otro is not None and otro.id != entidad.id:
            raise ValueError(f"Ya existe un género llamado {entidad.nombre}.")
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina un género que no tenga libros asociados."""
        if any(libro.genero.id == id for libro in self.__libros.leer_todos()):
            raise ValueError(
                "No se puede eliminar el género porque hay libros asociados."
            )
        return self.__repositorio.eliminar(id)


class EditorialServicio:
    """Reglas de alta, lectura, modificación y borrado de editoriales."""

    def __init__(
        self, repositorio: RepositorioEditorial, libros: RepositorioLibro
    ) -> None:
        self.__repositorio = repositorio
        self.__libros = libros

    def crear(self, entidad: Editorial) -> Editorial:
        """Crea una editorial si el nombre no está repetido."""
        if self.buscar_por_nombre(entidad.nombre) is not None:
            raise ValueError(f"Ya existe la editorial {entidad.nombre}.")
        return self.__repositorio.crear(entidad)

    def leer_por_id(self, id: int) -> Optional[Editorial]:
        """Lee una editorial por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[Editorial]:
        """Lee todas las editoriales."""
        return self.__repositorio.leer_todos()

    def buscar_por_nombre(self, nombre: str) -> Optional[Editorial]:
        """Busca una editorial por nombre."""
        for editorial in self.__repositorio.leer_todos():
            if _mismo_texto(editorial.nombre, nombre):
                return editorial
        return None

    def actualizar(self, entidad: Editorial) -> Editorial:
        """Actualiza una editorial."""
        if self.__repositorio.leer_por_id(entidad.id) is None:
            raise ValueError("No se encuentra la editorial para actualizar.")
        otro = self.buscar_por_nombre(entidad.nombre)
        if otro is not None and otro.id != entidad.id:
            raise ValueError(f"Ya existe la editorial {entidad.nombre}.")
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina una editorial que no tenga libros asociados."""
        if any(libro.editorial.id == id for libro in self.__libros.leer_todos()):
            raise ValueError(
                "No se puede eliminar la editorial porque hay libros asociados."
            )
        return self.__repositorio.eliminar(id)


class MonedaServicio:
    """Reglas de alta, lectura, modificación y borrado de monedas."""

    def __init__(self, repositorio: RepositorioMoneda, precios: RepositorioPrecio) -> None:
        self.__repositorio = repositorio
        self.__precios = precios

    def crear(self, entidad: Moneda) -> Moneda:
        """Crea una moneda si el código no está repetido."""
        if self.buscar_por_codigo(entidad.codigo) is not None:
            raise ValueError(f"Ya existe la moneda {entidad.codigo}.")
        return self.__repositorio.crear(entidad)

    def leer_por_id(self, id: int) -> Optional[Moneda]:
        """Lee una moneda por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[Moneda]:
        """Lee todas las monedas."""
        return self.__repositorio.leer_todos()

    def buscar_por_codigo(self, codigo: str) -> Optional[Moneda]:
        """Busca una moneda por su código."""
        buscado = codigo.strip().upper()
        for moneda in self.__repositorio.leer_todos():
            if moneda.codigo == buscado:
                return moneda
        return None

    def actualizar(self, entidad: Moneda) -> Moneda:
        """Actualiza una moneda."""
        if self.__repositorio.leer_por_id(entidad.id) is None:
            raise ValueError("No se encuentra la moneda para actualizar.")
        otro = self.buscar_por_codigo(entidad.codigo)
        if otro is not None and otro.id != entidad.id:
            raise ValueError(f"Ya existe la moneda {entidad.codigo}.")
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina una moneda que no esté usada en precios."""
        if any(precio.moneda.id == id for precio in self.__precios.leer_todos()):
            raise ValueError(
                "No se puede eliminar la moneda porque hay precios asociados."
            )
        return self.__repositorio.eliminar(id)


class TipoCotizacionServicio:
    """Reglas de alta, lectura, modificación y borrado de tipos de cotización."""

    def __init__(
        self,
        repositorio: RepositorioTipoCotizacion,
        cotizaciones: RepositorioCotizacionDolar,
    ) -> None:
        self.__repositorio = repositorio
        self.__cotizaciones = cotizaciones

    def crear(self, entidad: TipoCotizacion) -> TipoCotizacion:
        """Crea un tipo de cotización si el nombre no está repetido."""
        if self.buscar_por_nombre(entidad.nombre) is not None:
            raise ValueError(f"Ya existe el tipo de cotización {entidad.nombre}.")
        return self.__repositorio.crear(entidad)

    def leer_por_id(self, id: int) -> Optional[TipoCotizacion]:
        """Lee un tipo de cotización por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[TipoCotizacion]:
        """Lee todos los tipos de cotización."""
        return self.__repositorio.leer_todos()

    def buscar_por_nombre(self, nombre: str) -> Optional[TipoCotizacion]:
        """Busca un tipo de cotización por nombre."""
        for tipo in self.__repositorio.leer_todos():
            if _mismo_texto(tipo.nombre, nombre):
                return tipo
        return None

    def actualizar(self, entidad: TipoCotizacion) -> TipoCotizacion:
        """Actualiza un tipo de cotización."""
        if self.__repositorio.leer_por_id(entidad.id) is None:
            raise ValueError("No se encuentra el tipo de cotización para actualizar.")
        otro = self.buscar_por_nombre(entidad.nombre)
        if otro is not None and otro.id != entidad.id:
            raise ValueError(f"Ya existe el tipo de cotización {entidad.nombre}.")
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina un tipo que no tenga cotizaciones cargadas."""
        if self.__cotizaciones.leer_historico_por_tipo(id):
            raise ValueError(
                "No se puede eliminar el tipo porque tiene cotizaciones cargadas."
            )
        return self.__repositorio.eliminar(id)


class LibroServicio:
    """Reglas de alta, lectura, modificación y borrado de libros."""

    def __init__(
        self,
        repositorio: RepositorioLibro,
        generos: RepositorioGenero,
        editoriales: RepositorioEditorial,
        precios: RepositorioPrecio,
        stocks: RepositorioStock,
    ) -> None:
        self.__repositorio = repositorio
        self.__generos = generos
        self.__editoriales = editoriales
        self.__precios = precios
        self.__stocks = stocks

    def crear(self, entidad: Libro) -> Libro:
        """Crea un libro si el ISBN es único y las relaciones existen."""
        self.__validar_relaciones(entidad)
        if self.buscar_por_isbn(entidad.isbn) is not None:
            raise ValueError(f"Ya existe un libro con ISBN {entidad.isbn}.")
        return self.__repositorio.crear(entidad)

    def leer_por_id(self, id: int) -> Optional[Libro]:
        """Lee un libro por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[Libro]:
        """Lee todos los libros."""
        return self.__repositorio.leer_todos()

    def buscar_por_isbn(self, isbn: str) -> Optional[Libro]:
        """Busca un libro por ISBN."""
        for libro in self.__repositorio.leer_todos():
            if _mismo_texto(libro.isbn, isbn):
                return libro
        return None

    def actualizar(self, entidad: Libro) -> Libro:
        """Actualiza un libro."""
        if self.__repositorio.leer_por_id(entidad.id) is None:
            raise ValueError("No se encuentra el libro para actualizar.")
        self.__validar_relaciones(entidad)
        otro = self.buscar_por_isbn(entidad.isbn)
        if otro is not None and otro.id != entidad.id:
            raise ValueError(f"Ya existe un libro con ISBN {entidad.isbn}.")
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina un libro que no tenga precio ni stock."""
        if any(precio.libro.id == id for precio in self.__precios.leer_todos()):
            raise ValueError(
                "No se puede eliminar el libro porque tiene precios asociados."
            )
        if self.__stocks.leer_por_libro(id) is not None:
            raise ValueError(
                "No se puede eliminar el libro porque tiene stock asociado."
            )
        return self.__repositorio.eliminar(id)

    def __validar_relaciones(self, entidad: Libro) -> None:
        if self.__generos.leer_por_id(entidad.genero.id) is None:
            raise ValueError("El género indicado no existe.")
        if self.__editoriales.leer_por_id(entidad.editorial.id) is None:
            raise ValueError("La editorial indicada no existe.")


class PrecioServicio:
    """Reglas de alta, lectura, modificación y borrado de precios."""

    def __init__(
        self,
        repositorio: RepositorioPrecio,
        libros: RepositorioLibro,
        monedas: RepositorioMoneda,
    ) -> None:
        self.__repositorio = repositorio
        self.__libros = libros
        self.__monedas = monedas

    def crear(self, entidad: Precio) -> Precio:
        """Crea un precio si el libro y la moneda existen y no hay duplicado."""
        self.__validar_relaciones(entidad)
        if self.buscar(entidad.libro.id, entidad.moneda.id) is not None:
            raise ValueError("Ya existe un precio para ese libro y esa moneda.")
        return self.__repositorio.crear(entidad)

    def leer_por_id(self, id: int) -> Optional[Precio]:
        """Lee un precio por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[Precio]:
        """Lee todos los precios."""
        return self.__repositorio.leer_todos()

    def buscar(self, libro_id: int, moneda_id: int) -> Optional[Precio]:
        """Busca el precio de un libro en una moneda."""
        for precio in self.__repositorio.leer_todos():
            if precio.libro.id == libro_id and precio.moneda.id == moneda_id:
                return precio
        return None

    def actualizar(self, entidad: Precio) -> Precio:
        """Actualiza un precio."""
        if self.__repositorio.leer_por_id(entidad.id) is None:
            raise ValueError("No se encuentra el precio para actualizar.")
        self.__validar_relaciones(entidad)
        otro = self.buscar(entidad.libro.id, entidad.moneda.id)
        if otro is not None and otro.id != entidad.id:
            raise ValueError("Ya existe un precio para ese libro y esa moneda.")
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina un precio por id."""
        return self.__repositorio.eliminar(id)

    def __validar_relaciones(self, entidad: Precio) -> None:
        if self.__libros.leer_por_id(entidad.libro.id) is None:
            raise ValueError("El libro indicado no existe.")
        if self.__monedas.leer_por_id(entidad.moneda.id) is None:
            raise ValueError("La moneda indicada no existe.")


class StockServicio:
    """Reglas de alta, lectura, modificación y borrado de stock."""

    def __init__(self, repositorio: RepositorioStock, libros: RepositorioLibro) -> None:
        self.__repositorio = repositorio
        self.__libros = libros

    def crear(self, stock: Stock) -> Stock:
        """Crea el stock de un libro."""
        self.__validar_libro(stock.libro.id)
        if self.__repositorio.leer_por_libro(stock.libro.id) is not None:
            raise ValueError("Ya existe stock para ese libro.")
        return self.__repositorio.crear(stock)

    def leer_por_libro(self, libro_id: int) -> Optional[Stock]:
        """Lee el stock de un libro."""
        return self.__repositorio.leer_por_libro(libro_id)

    def leer_todos(self) -> List[Stock]:
        """Lee todos los stocks."""
        return self.__repositorio.leer_todos()

    def actualizar(self, stock: Stock) -> Stock:
        """Actualiza el stock de un libro."""
        self.__validar_libro(stock.libro.id)
        return self.__repositorio.actualizar(stock)

    def eliminar(self, libro_id: int) -> bool:
        """Elimina el stock de un libro."""
        return self.__repositorio.eliminar(libro_id)

    def __validar_libro(self, libro_id: int) -> None:
        if self.__libros.leer_por_id(libro_id) is None:
            raise ValueError("El libro indicado no existe.")


class ServicioCotizacionDolar:
    """Reglas de cotización y descarga desde la API del dólar."""

    def __init__(
        self,
        repositorio: RepositorioCotizacionDolar,
        tipos: RepositorioTipoCotizacion,
    ) -> None:
        self.__repositorio = repositorio
        self.__tipos = tipos

    def crear(self, cotizacion: CotizacionDolar) -> CotizacionDolar:
        """Crea una cotización si el tipo existe y la fecha no está repetida."""
        self.__validar_tipo(cotizacion.tipo_cotizacion.id)
        existente = self.__repositorio.leer_por_tipo_y_fecha(
            cotizacion.tipo_cotizacion.id,
            cotizacion.fecha,
        )
        if existente is not None:
            raise ValueError("Ya existe una cotización para ese tipo y esa fecha.")
        return self.__repositorio.crear(cotizacion)

    def leer_por_tipo_y_fecha(
        self, tipo_id: int, fecha: date
    ) -> Optional[CotizacionDolar]:
        """Lee una cotización por tipo y fecha."""
        return self.__repositorio.leer_por_tipo_y_fecha(tipo_id, fecha)

    def leer_historico_por_tipo(self, tipo_id: int) -> List[CotizacionDolar]:
        """Lee el histórico de un tipo de cotización."""
        self.__validar_tipo(tipo_id)
        return self.__repositorio.leer_historico_por_tipo(tipo_id)

    def leer_todas(self) -> List[CotizacionDolar]:
        """Lee todas las cotizaciones."""
        return self.__repositorio.leer_todas()

    def actualizar(self, cotizacion: CotizacionDolar) -> CotizacionDolar:
        """Actualiza compra y venta de una cotización existente."""
        self.__validar_tipo(cotizacion.tipo_cotizacion.id)
        return self.__repositorio.actualizar(cotizacion)

    def eliminar(self, tipo_id: int, fecha: date) -> bool:
        """Elimina una cotización por tipo y fecha."""
        return self.__repositorio.eliminar(tipo_id, fecha)

    def obtener_cotizaciones(self) -> List[CotizacionDolar]:
        """Descarga las cotizaciones de la API y las registra en la base.

        La URL se lee de la variable API_URL del archivo .env.

        Returns:
            List[CotizacionDolar]: Cotizaciones creadas o actualizadas.

        Raises:
            ValueError: Si falta API_URL, la respuesta no es una lista
                o no hay conexión con la API (URLError).
        """
        cargar_env()
        url = os.getenv("API_URL", "").strip()
        if url == "":
            raise ValueError("Falta API_URL en el archivo .env.")
        pedido = Request(
            url,
            headers={"User-Agent": "BookManager/1.0 (Grupo 28; educational)"},
        )
        try:
            with urlopen(pedido, timeout=30) as respuesta:
                payload = json.loads(respuesta.read().decode("utf-8"))
        except URLError as error:
            raise ValueError(
                "No se pudo conectar con la API del dólar. "
                f"Revisá la red o API_URL. Detalle: {error.reason}"
            ) from error
        if not isinstance(payload, list):
            raise ValueError("La API del dólar no devolvió una lista.")
        registradas: List[CotizacionDolar] = []
        for item in payload:
            casa = str(item.get("casa", "")).strip().lower()
            nombre = MAPA_CASA_DOLAR.get(casa)
            if nombre is None:
                continue
            compra = item.get("compra")
            venta = item.get("venta")
            if compra is None or venta is None:
                continue
            if float(compra) <= 0 or float(venta) <= 0 or float(compra) > float(venta):
                continue
            fecha = _fecha_de_api(item.get("fechaActualizacion"))
            tipo = self.__tipo_para(nombre, str(item.get("nombre") or nombre))
            cotizacion = CotizacionDolar(
                tipo_cotizacion=tipo,
                fecha=fecha,
                compra=float(compra),
                venta=float(venta),
            )
            existente = self.__repositorio.leer_por_tipo_y_fecha(tipo.id, fecha)
            if existente is None:
                registradas.append(self.__repositorio.crear(cotizacion))
            else:
                registradas.append(self.__repositorio.actualizar(cotizacion))
        return registradas

    def __tipo_para(self, nombre: str, descripcion: str) -> TipoCotizacion:
        existente = None
        for tipo in self.__tipos.leer_todos():
            if _mismo_texto(tipo.nombre, nombre):
                existente = tipo
                break
        if existente is not None:
            return existente
        return self.__tipos.crear(
            TipoCotizacion(id=0, nombre=nombre, descripcion=descripcion)
        )

    def __validar_tipo(self, tipo_id: int) -> None:
        if self.__tipos.leer_por_id(tipo_id) is None:
            raise ValueError("El tipo de cotización indicado no existe.")


def _fecha_de_api(valor: Optional[str]) -> date:
    """Toma la fecha de la marca de tiempo que devuelve la API."""
    if not valor:
        return date.today()
    texto = str(valor).replace("Z", "")
    try:
        return datetime.fromisoformat(texto).date()
    except ValueError:
        return date.fromisoformat(texto[:10])


@dataclass(frozen=True)
class ComparacionCuspide:
    """Comparación entre el precio público de Cúspide y el costo de reposición."""

    titulo: str
    autor: str
    precio_cuspide_ars: float
    costo_usd: float
    tipo_cotizacion: str
    cotizacion_venta: float
    costo_reposicion_ars: float

    @property
    def diferencia_ars(self) -> float:
        """Diferencia entre el precio de Cúspide y el costo en pesos."""
        return round(self.precio_cuspide_ars - self.costo_reposicion_ars, 2)

    @property
    def cubre_el_costo(self) -> bool:
        """Indica si el precio de Cúspide alcanza para cubrir la reposición."""
        return self.diferencia_ars >= 0


@dataclass(frozen=True)
class LineaStock:
    """Fila del reporte de reposición."""

    titulo: str
    autor: str
    cantidad: int
    punto_reposicion: int


@dataclass(frozen=True)
class LineaHistorico:
    """Fila del histórico de una cotización."""

    fecha: date
    tipo: str
    compra: float
    venta: float


@dataclass(frozen=True)
class LineaBimonetaria:
    """Precio de un libro en pesos y en otra moneda."""

    titulo: str
    autor: str
    precio_ars: Optional[float]
    codigo_moneda: str
    precio_otra: Optional[float]


class ReporteServicio:
    """Reportes de cotización, Cúspide, stock, histórico y precios bimonetarios."""

    def __init__(
        self,
        libros: RepositorioLibro,
        precios: RepositorioPrecio,
        monedas: RepositorioMoneda,
        stocks: RepositorioStock,
        cotizaciones: RepositorioCotizacionDolar,
        tipos: RepositorioTipoCotizacion,
    ) -> None:
        self.__libros = libros
        self.__precios = precios
        self.__monedas = monedas
        self.__stocks = stocks
        self.__cotizaciones = cotizaciones
        self.__tipos = tipos

    def comparar_con_cuspide(self, tipo_id: int) -> List[ComparacionCuspide]:
        """Cotiza el costo en dólares y lo compara con el precio de Cúspide."""
        tipo = self.__tipos.leer_por_id(tipo_id)
        if tipo is None:
            raise ValueError("El tipo de cotización indicado no existe.")
        historico = self.__cotizaciones.leer_historico_por_tipo(tipo_id)
        if not historico:
            raise ValueError("No hay cotizaciones para el tipo indicado.")
        cotizacion = historico[-1]
        moneda_ars = self.__moneda_por_codigo("ARS")
        moneda_usd = self.__moneda_por_codigo("USD")
        filas: List[ComparacionCuspide] = []
        for libro in self.__libros.leer_todos():
            precio_ars = self.__importe(libro.id, moneda_ars.id)
            costo_usd = self.__importe(libro.id, moneda_usd.id)
            if precio_ars is None or costo_usd is None:
                continue
            costo_ars = round(costo_usd * cotizacion.venta, 2)
            filas.append(
                ComparacionCuspide(
                    titulo=libro.titulo,
                    autor=libro.autor,
                    precio_cuspide_ars=precio_ars,
                    costo_usd=costo_usd,
                    tipo_cotizacion=tipo.nombre,
                    cotizacion_venta=cotizacion.venta,
                    costo_reposicion_ars=costo_ars,
                )
            )
        return filas

    def stock_a_reponer(self) -> List[LineaStock]:
        """Devuelve los libros con stock en el punto de reposición o por debajo."""
        filas: List[LineaStock] = []
        for stock in self.__stocks.leer_todos():
            if stock.cantidad > stock.punto_reposicion:
                continue
            filas.append(
                LineaStock(
                    titulo=stock.libro.titulo,
                    autor=stock.libro.autor,
                    cantidad=stock.cantidad,
                    punto_reposicion=stock.punto_reposicion,
                )
            )
        return filas

    def historico(self, tipo_id: int) -> List[LineaHistorico]:
        """Devuelve el histórico de un tipo de dólar, ordenado por fecha."""
        tipo = self.__tipos.leer_por_id(tipo_id)
        if tipo is None:
            raise ValueError("El tipo de cotización indicado no existe.")
        return [
            LineaHistorico(
                fecha=item.fecha,
                tipo=tipo.nombre,
                compra=item.compra,
                venta=item.venta,
            )
            for item in self.__cotizaciones.leer_historico_por_tipo(tipo_id)
        ]

    def lista_bimonetaria(self, codigo_moneda: str) -> List[LineaBimonetaria]:
        """Lista cada libro en pesos y en la moneda elegida.

        Args:
            codigo_moneda (str): Código de tres letras de la segunda moneda.

        Returns:
            List[LineaBimonetaria]: Una fila por libro.
        """
        moneda_ars = self.__moneda_por_codigo("ARS")
        otra = self.__moneda_por_codigo(codigo_moneda)
        filas: List[LineaBimonetaria] = []
        for libro in self.__libros.leer_todos():
            filas.append(
                LineaBimonetaria(
                    titulo=libro.titulo,
                    autor=libro.autor,
                    precio_ars=self.__importe(libro.id, moneda_ars.id),
                    codigo_moneda=otra.codigo,
                    precio_otra=self.__importe(libro.id, otra.id),
                )
            )
        return filas

    def exportar_precios_csv(self, ruta: Optional[Path] = None) -> Path:
        """Exporta el precio de cada libro en todas las monedas cargadas.

        Args:
            ruta (Optional[Path]): Destino del CSV. Si no se indica, se usa
                book_manager/export/precios.csv.

        Returns:
            Path: Ruta del archivo generado.
        """
        if ruta is None:
            ruta = Path(__file__).resolve().parents[3] / "export" / "precios.csv"
        ruta.parent.mkdir(parents=True, exist_ok=True)
        with ruta.open("w", encoding="utf-8", newline="") as archivo:
            escritor = csv.DictWriter(
                archivo,
                fieldnames=["isbn", "titulo", "autor", "moneda", "importe"],
            )
            escritor.writeheader()
            for precio in self.__precios.leer_todos():
                escritor.writerow(
                    {
                        "isbn": precio.libro.isbn,
                        "titulo": precio.libro.titulo,
                        "autor": precio.libro.autor,
                        "moneda": precio.moneda.codigo,
                        "importe": f"{precio.importe:.2f}",
                    }
                )
        return ruta

    def __moneda_por_codigo(self, codigo: str) -> Moneda:
        buscado = codigo.strip().upper()
        for moneda in self.__monedas.leer_todos():
            if moneda.codigo == buscado:
                return moneda
        raise ValueError(f"No está cargada la moneda {codigo}.")

    def __importe(self, libro_id: int, moneda_id: int) -> Optional[float]:
        for precio in self.__precios.leer_todos():
            if precio.libro.id == libro_id and precio.moneda.id == moneda_id:
                return precio.importe
        return None


@dataclass(frozen=True)
class LineaPrecioWeb:
    """Comparación entre el precio interno y el precio publicado en Cúspide."""

    libro: str
    precio_interno: Optional[float]
    precio_web: Optional[float]
    diferencia: Optional[float]
    fecha_extraccion: Optional[datetime]


def armar_comparaciones(
    libros: RepositorioLibro,
    precios: RepositorioPrecio,
    extracciones: RepositorioExtraccionWeb,
) -> List[LineaPrecioWeb]:
    """Arma una fila por libro del catálogo.

    Usa la extracción cuyo título coincide con el del catálogo.
    """
    por_libro: dict[int, List] = {}
    for fila in extracciones.leer_todos():
        por_libro.setdefault(fila.libro_id, []).append(fila)
    resultado = []
    for libro in libros.leer_todos():
        interno = _importe_de(precios, libro.id, "ARS")
        candidatas = por_libro.get(libro.id, [])
        elegida = _elegir_extraccion(candidatas, libro.titulo, interno)
        web = elegida.precio_web if elegida is not None else None
        fecha = elegida.fecha_extraccion if elegida is not None else None
        diferencia = None
        if interno is not None and web is not None:
            diferencia = round(web - interno, 2)
        resultado.append(
            LineaPrecioWeb(
                libro=f"{libro.titulo} ({libro.autor})",
                precio_interno=interno,
                precio_web=web,
                diferencia=diferencia,
                fecha_extraccion=fecha,
            )
        )
    return resultado


def _elegir_extraccion(candidatas, titulo: str, interno: Optional[float]):
    """Prefiere el título exacto y, si hay varios, el precio más cercano."""
    buscado = " ".join(titulo.split()).casefold()
    exactas = [
        fila
        for fila in candidatas
        if " ".join(fila.titulo_web.split()).casefold() == buscado
    ]
    pool = exactas or [fila for fila in candidatas if fila.coincide]
    if not pool:
        return None

    def distancia(fila) -> float:
        if fila.precio_web is None or interno is None:
            return 10**12
        return abs(fila.precio_web - interno)

    return min(pool, key=distancia)


def _importe_de(
    precios: RepositorioPrecio,
    libro_id: int,
    codigo: str,
) -> Optional[float]:
    for precio in precios.leer_todos():
        if precio.libro.id == libro_id and precio.moneda.codigo == codigo:
            return precio.importe
    return None


class ServicioScraping:
    """Ejecuta el spider y deja el CSV de alertas."""

    def __init__(
        self,
        extracciones: RepositorioExtraccionWeb,
        libros: RepositorioLibro,
        precios: RepositorioPrecio,
    ) -> None:
        self.__extracciones = extracciones
        self.__libros = libros
        self.__precios = precios

    def ejecutar(self, umbral: float, limite: int = 0) -> Path:
        """Pide el umbral, corre el scraper y escribe las alertas.

        Args:
            umbral (float): Diferencia mínima, en pesos, para alertar.
            limite (int): Títulos a buscar. Cero recorre el catálogo.

        Returns:
            Path: CSV de alertas.

        Raises:
            ValueError: Si el umbral es negativo o el scraper falla.
        """
        if umbral < 0:
            raise ValueError("La diferencia tiene que ser mayor o igual a cero.")
        self._correr(limite)
        return self.generar_alertas_csv(umbral)

    def generar_alertas_csv(self, umbral: float) -> Path:
        """Escribe el CSV descargable con los libros que superan el umbral.

        Args:
            umbral (float): Diferencia mínima en pesos.

        Returns:
            Path: Ruta del CSV.
        """
        if umbral < 0:
            raise ValueError("La diferencia tiene que ser mayor o igual a cero.")
        carpeta = Path(__file__).resolve().parents[1] / "reportes"
        carpeta.mkdir(parents=True, exist_ok=True)
        destino = carpeta / "alertas.csv"
        filas = [
            fila
            for fila in armar_comparaciones(
                self.__libros, self.__precios, self.__extracciones
            )
            if fila.diferencia is not None and abs(fila.diferencia) >= umbral
        ]
        with destino.open("w", encoding="utf-8", newline="") as archivo:
            escritor = csv.DictWriter(
                archivo,
                fieldnames=[
                    "libro",
                    "precio_interno",
                    "precio_web",
                    "diferencia",
                    "umbral",
                    "fecha_extraccion",
                ],
            )
            escritor.writeheader()
            for fila in filas:
                fecha = (
                    fila.fecha_extraccion.strftime("%Y-%m-%d %H:%M:%S")
                    if fila.fecha_extraccion is not None
                    else ""
                )
                escritor.writerow(
                    {
                        "libro": fila.libro,
                        "precio_interno": fila.precio_interno,
                        "precio_web": fila.precio_web,
                        "diferencia": fila.diferencia,
                        "umbral": umbral,
                        "fecha_extraccion": fecha,
                    }
                )
        return destino

    def generar_reporte_excel(self) -> Path:
        """Genera el Excel de precios internos y web.

        Returns:
            Path: Ruta del archivo .xlsx.
        """
        from openpyxl import Workbook

        carpeta = Path(__file__).resolve().parents[1] / "reportes"
        carpeta.mkdir(parents=True, exist_ok=True)
        destino = carpeta / "comparacion_precios.xlsx"
        libro = Workbook()
        hoja = libro.active
        hoja.title = "Precios"
        hoja.append(
            [
                "Libro",
                "Precio interno",
                "Precio web",
                "Diferencia",
                "Fecha de extracción",
            ]
        )
        for fila in armar_comparaciones(
            self.__libros, self.__precios, self.__extracciones
        ):
            fecha = (
                fila.fecha_extraccion.strftime("%Y-%m-%d %H:%M:%S")
                if fila.fecha_extraccion is not None
                else ""
            )
            hoja.append(
                [
                    fila.libro,
                    fila.precio_interno,
                    fila.precio_web,
                    fila.diferencia,
                    fecha,
                ]
            )
        libro.save(destino)
        return destino

    def _correr(self, limite: int) -> None:
        entorno = os.environ.copy()
        entorno["PYTHONPATH"] = str(Path(__file__).resolve().parents[2])
        resultado = subprocess.run(
            [sys.executable, "-m", "book_manager.scraper.ejecutar", str(limite)],
            env=entorno,
            cwd=str(Path(__file__).resolve().parents[4]),
        )
        if resultado.returncode != 0:
            raise ValueError("El scraper de Cúspide no pudo completar la búsqueda.")


class ServicioAuditoria:
    """Lectura y alta del historial de auditoría."""

    def __init__(self, repositorio: RepositorioAuditoria) -> None:
        self.__repositorio = repositorio

    def crear(self, accion: str, detalles: str) -> Auditoria:
        """Registra una acción."""
        return self.__repositorio.crear(
            Auditoria(
                id=0,
                accion=accion,
                fecha=datetime.now(),
                detalles=detalles,
            )
        )

    def leer_por_id(self, id: int) -> Optional[Auditoria]:
        """Lee un registro por id."""
        return self.__repositorio.leer_por_id(id)

    def leer_todos(self) -> List[Auditoria]:
        """Devuelve el historial completo."""
        return self.__repositorio.leer_todos()

    def actualizar(self, entidad: Auditoria) -> Auditoria:
        """Actualiza un registro."""
        return self.__repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        """Elimina un registro."""
        return self.__repositorio.eliminar(id)


def _registrar_auditoria(accion: str, detalles: str) -> None:
    """Guarda la auditoría en su propia transacción."""
    import book_manager.models.models  # registra la tabla

    db = ConexionDB()
    db.crear_tablas()
    RepositorioAuditoria(db).crear(
        Auditoria(
            id=0,
            accion=accion[:80],
            fecha=datetime.now(),
            detalles=detalles[:500],
        )
    )


def _resumir(valor: object) -> str:
    if isinstance(valor, list):
        return f"{len(valor)} filas"
    texto = str(valor)
    if len(texto) > 160:
        return texto[:157] + "..."
    return texto


def auditar(accion: str):
    """Decora un método para dejar constancia de la operación.

    Args:
        accion (str): Nombre de la operación que se guarda en la tabla.
    """

    def decorador(metodo):
        @wraps(metodo)
        def envoltura(self, *args, **kwargs):
            try:
                resultado = metodo(self, *args, **kwargs)
            except Exception as error:
                _registrar_auditoria(accion, f"{metodo.__name__} error: {error}")
                raise
            _registrar_auditoria(
                accion,
                f"{metodo.__name__} -> {_resumir(resultado)}",
            )
            return resultado

        return envoltura

    return decorador


def auditar_operaciones(clase):
    """Aplica el decorador a cada operación pública de la clase."""
    for nombre, metodo in list(clase.__dict__.items()):
        if nombre.startswith("_") or not isinstance(metodo, FunctionType):
            continue
        setattr(clase, nombre, auditar(nombre)(metodo))
    return clase


for _clase in (
    GeneroServicio,
    EditorialServicio,
    MonedaServicio,
    TipoCotizacionServicio,
    LibroServicio,
    PrecioServicio,
    StockServicio,
    ServicioCotizacionDolar,
    ReporteServicio,
    ServicioScraping,
):
    auditar_operaciones(_clase)


@dataclass
class Contexto:
    """Servicios que usa la consola y la demostración."""

    generos: GeneroServicio
    editoriales: EditorialServicio
    monedas: MonedaServicio
    tipos_cotizacion: TipoCotizacionServicio
    libros: LibroServicio
    precios: PrecioServicio
    stocks: StockServicio
    cotizaciones: ServicioCotizacionDolar
    reportes: ReporteServicio
    scraping: ServicioScraping
    auditoria: ServicioAuditoria


ServicioGenero = GeneroServicio
ServicioEditorial = EditorialServicio
ServicioMoneda = MonedaServicio
ServicioTipoCotizacion = TipoCotizacionServicio
ServicioLibro = LibroServicio
ServicioStock = StockServicio
''', encoding='utf-8')
print('Vista —', destino.name, destino.stat().st_size, 'bytes')
````

````python
#@title Ejercicio 07 — menú de scraping, reporte y auditoría
from pathlib import Path
destino = Path(RUTA_SRC) / 'book_manager/ui/console.py'
destino.parent.mkdir(parents=True, exist_ok=True)
destino.write_text(r'''"""Consola de Book Manager.

Toda la entrada y la salida pasan por acá. Las reglas quedan en los
servicios. La demostración cubre el listado y el alta o la modificación
que pide el notebook, sin esperar teclas, para que "Ejecutar todo" pueda
seguir con las celdas de git.
"""

from dataclasses import dataclass
from datetime import date
from pathlib import Path
from typing import Dict
from typing import Optional
from typing import Sequence
from urllib.error import URLError

from book_manager.entities.entities import CotizacionDolar
from book_manager.entities.entities import Editorial
from book_manager.entities.entities import Genero
from book_manager.entities.entities import Libro
from book_manager.entities.entities import Moneda
from book_manager.entities.entities import Precio
from book_manager.entities.entities import Stock
from book_manager.entities.entities import TipoCotizacion
from book_manager.services.services import ComparacionCuspide
from book_manager.services.services import Contexto
from book_manager.services.services import LineaBimonetaria
from book_manager.services.services import LineaHistorico
from book_manager.services.services import LineaStock


def formatear_pesos(valor: float) -> str:
    """Formatea un importe con separador de miles y coma decimal."""
    texto = f"{valor:,.2f}"
    return "$ " + texto.replace(",", "X").replace(".", ",").replace("X", ".")


@dataclass
class _InformeGuardado:
    """Copia de un reporte emitido durante la sesión de consola."""

    id: int
    titulo: str
    detalle: str


class Consola:
    """Menú de consola con el CRUD de cada clase y los reportes."""

    def __init__(self, contexto: Contexto) -> None:
        self.__contexto = contexto
        self.__informes: Dict[int, _InformeGuardado] = {}
        self.__proximo_informe = 1

    def ejecutar(self) -> None:
        """Abre el menú principal hasta que se elige salir."""
        try:
            self.__bucle_principal()
        except EOFError:
            print("\nSe cerró la entrada. Fin de la sesión.")

    def __bucle_principal(self) -> None:
        while True:
            print("\n=== Book Manager - Sprint 3 ===")
            print("Referencia de catálogo: https://cuspide.com/")
            print("1. Géneros")
            print("2. Editoriales")
            print("3. Monedas")
            print("4. Tipos de cotización")
            print("5. Libros")
            print("6. Precios")
            print("7. Stock")
            print("8. Cotizaciones del dólar")
            print("9. Reportes")
            print("10. Obtener las cotizaciones por API")
            print("11. Lista de precios bimonetaria")
            print("12. Exportar los precios a CSV")
            print("13. Ejecutar scraping")
            print("14. Generar reporte")
            print("15. Ver historial de auditoría")
            print("0. Salir")
            opcion = input("Opción: ").strip()
            if opcion == "0":
                print("Fin de la sesión.")
                return
            try:
                match opcion:
                    case "1":
                        self.__menu_generos()
                    case "2":
                        self.__menu_editoriales()
                    case "3":
                        self.__menu_monedas()
                    case "4":
                        self.__menu_tipos()
                    case "5":
                        self.__menu_libros()
                    case "6":
                        self.__menu_precios()
                    case "7":
                        self.__menu_stock()
                    case "8":
                        self.__menu_cotizaciones()
                    case "9":
                        self.__menu_reportes()
                    case "10":
                        self.__obtener_cotizaciones_api()
                    case "11":
                        self.__lista_bimonetaria()
                    case "12":
                        ruta = self.__contexto.reportes.exportar_precios_csv()
                        print(f"Precios exportados en {ruta}")
                    case "13":
                        self.__ejecutar_scraping()
                    case "14":
                        ruta = self.__contexto.scraping.generar_reporte_excel()
                        print(f"Reporte Excel generado en {ruta}")
                    case "15":
                        self.__ver_auditoria()
                    case _:
                        print("Opción inválida.")
            except (ValueError, URLError, OSError) as error:
                print(f"No se pudo completar la operación: {error}")

    def __menu_generos(self) -> None:
        servicio = self.__contexto.generos
        while True:
            self.__titulo("Géneros")
            opcion = self.__leer_opcion_crud()
            if opcion == "0":
                return
            match opcion:
                case "1":
                    self.__listar(servicio.leer_todos())
                case "2":
                    creado = servicio.crear(
                        Genero(
                            id=0,
                            nombre=self.__pedir_texto("Nombre: "),
                            descripcion=self.__pedir_texto("Descripción: "),
                        )
                    )
                    print(f"Alta realizada: {creado}")
                case "3":
                    actual = self.__exigir(servicio.leer_por_id(self.__pedir_entero("Id: ")))
                    actual.nombre = self.__pedir_texto(
                        f"Nombre [{actual.nombre}]: ", actual.nombre
                    )
                    actual.descripcion = self.__pedir_texto(
                        f"Descripción [{actual.descripcion}]: ",
                        actual.descripcion,
                    )
                    print(f"Modificación realizada: {servicio.actualizar(actual)}")
                case "4":
                    self.__avisar_borrado(servicio.eliminar(self.__pedir_entero("Id: ")))
                case "5":
                    self.__mostrar(servicio.leer_por_id(self.__pedir_entero("Id: ")))
                case _:
                    print("Opción inválida.")

    def __menu_editoriales(self) -> None:
        servicio = self.__contexto.editoriales
        while True:
            self.__titulo("Editoriales")
            opcion = self.__leer_opcion_crud()
            if opcion == "0":
                return
            match opcion:
                case "1":
                    self.__listar(servicio.leer_todos())
                case "2":
                    creado = servicio.crear(
                        Editorial(
                            id=0,
                            nombre=self.__pedir_texto("Nombre: "),
                            pais=self.__pedir_texto("País: "),
                            sitio_web=self.__pedir_texto("Sitio web: "),
                        )
                    )
                    print(f"Alta realizada: {creado}")
                case "3":
                    actual = self.__exigir(servicio.leer_por_id(self.__pedir_entero("Id: ")))
                    actual.nombre = self.__pedir_texto(
                        f"Nombre [{actual.nombre}]: ", actual.nombre
                    )
                    actual.pais = self.__pedir_texto(
                        f"País [{actual.pais}]: ", actual.pais
                    )
                    actual.sitio_web = self.__pedir_texto(
                        f"Sitio web [{actual.sitio_web}]: ", actual.sitio_web
                    )
                    print(f"Modificación realizada: {servicio.actualizar(actual)}")
                case "4":
                    self.__avisar_borrado(servicio.eliminar(self.__pedir_entero("Id: ")))
                case "5":
                    self.__mostrar(servicio.leer_por_id(self.__pedir_entero("Id: ")))
                case _:
                    print("Opción inválida.")

    def __menu_monedas(self) -> None:
        servicio = self.__contexto.monedas
        while True:
            self.__titulo("Monedas")
            opcion = self.__leer_opcion_crud()
            if opcion == "0":
                return
            match opcion:
                case "1":
                    self.__listar(servicio.leer_todos())
                case "2":
                    creado = servicio.crear(
                        Moneda(
                            id=0,
                            codigo=self.__pedir_texto("Código de 3 letras: "),
                            nombre=self.__pedir_texto("Nombre: "),
                            simbolo=self.__pedir_texto("Símbolo: "),
                        )
                    )
                    print(f"Alta realizada: {creado}")
                case "3":
                    actual = self.__exigir(servicio.leer_por_id(self.__pedir_entero("Id: ")))
                    actual.codigo = self.__pedir_texto(
                        f"Código [{actual.codigo}]: ", actual.codigo
                    )
                    actual.nombre = self.__pedir_texto(
                        f"Nombre [{actual.nombre}]: ", actual.nombre
                    )
                    actual.simbolo = self.__pedir_texto(
                        f"Símbolo [{actual.simbolo}]: ", actual.simbolo
                    )
                    print(f"Modificación realizada: {servicio.actualizar(actual)}")
                case "4":
                    self.__avisar_borrado(servicio.eliminar(self.__pedir_entero("Id: ")))
                case "5":
                    self.__mostrar(servicio.leer_por_id(self.__pedir_entero("Id: ")))
                case _:
                    print("Opción inválida.")

    def __menu_tipos(self) -> None:
        servicio = self.__contexto.tipos_cotizacion
        while True:
            self.__titulo("Tipos de cotización")
            opcion = self.__leer_opcion_crud()
            if opcion == "0":
                return
            match opcion:
                case "1":
                    self.__listar(servicio.leer_todos())
                case "2":
                    creado = servicio.crear(
                        TipoCotizacion(
                            id=0,
                            nombre=self.__pedir_texto("Nombre: "),
                            descripcion=self.__pedir_texto("Descripción: "),
                        )
                    )
                    print(f"Alta realizada: {creado}")
                case "3":
                    actual = self.__exigir(servicio.leer_por_id(self.__pedir_entero("Id: ")))
                    actual.nombre = self.__pedir_texto(
                        f"Nombre [{actual.nombre}]: ", actual.nombre
                    )
                    actual.descripcion = self.__pedir_texto(
                        f"Descripción [{actual.descripcion}]: ",
                        actual.descripcion,
                    )
                    print(f"Modificación realizada: {servicio.actualizar(actual)}")
                case "4":
                    self.__avisar_borrado(servicio.eliminar(self.__pedir_entero("Id: ")))
                case "5":
                    self.__mostrar(servicio.leer_por_id(self.__pedir_entero("Id: ")))
                case _:
                    print("Opción inválida.")

    def __menu_libros(self) -> None:
        servicio = self.__contexto.libros
        while True:
            self.__titulo("Libros")
            opcion = self.__leer_opcion_crud()
            if opcion == "0":
                return
            match opcion:
                case "1":
                    self.__listar(servicio.leer_todos())
                case "2":
                    print("Géneros:")
                    self.__listar(self.__contexto.generos.leer_todos())
                    print("Editoriales:")
                    self.__listar(self.__contexto.editoriales.leer_todos())
                    creado = servicio.crear(
                        Libro(
                            id=0,
                            isbn=self.__pedir_texto("ISBN: "),
                            titulo=self.__pedir_texto("Título: "),
                            autor=self.__pedir_texto("Autor: "),
                            editorial=self.__exigir(
                                self.__contexto.editoriales.leer_por_id(
                                    self.__pedir_entero("Id de editorial: ")
                                )
                            ),
                            genero=self.__exigir(
                                self.__contexto.generos.leer_por_id(
                                    self.__pedir_entero("Id de género: ")
                                )
                            ),
                            anio=self.__pedir_entero("Año: "),
                            paginas=self.__pedir_entero("Páginas: "),
                        )
                    )
                    print(f"Alta realizada: {creado}")
                case "3":
                    actual = self.__exigir(servicio.leer_por_id(self.__pedir_entero("Id: ")))
                    actual.isbn = self.__pedir_texto(f"ISBN [{actual.isbn}]: ", actual.isbn)
                    actual.titulo = self.__pedir_texto(
                        f"Título [{actual.titulo}]: ", actual.titulo
                    )
                    actual.autor = self.__pedir_texto(
                        f"Autor [{actual.autor}]: ", actual.autor
                    )
                    actual.editorial = self.__exigir(
                        self.__contexto.editoriales.leer_por_id(
                            self.__pedir_entero(
                                f"Id de editorial [{actual.editorial.id}]: ",
                                actual.editorial.id,
                            )
                        )
                    )
                    actual.genero = self.__exigir(
                        self.__contexto.generos.leer_por_id(
                            self.__pedir_entero(
                                f"Id de género [{actual.genero.id}]: ",
                                actual.genero.id,
                            )
                        )
                    )
                    actual.anio = self.__pedir_entero(f"Año [{actual.anio}]: ", actual.anio)
                    actual.paginas = self.__pedir_entero(
                        f"Páginas [{actual.paginas}]: ", actual.paginas
                    )
                    print(f"Modificación realizada: {servicio.actualizar(actual)}")
                case "4":
                    self.__avisar_borrado(servicio.eliminar(self.__pedir_entero("Id: ")))
                case "5":
                    self.__mostrar(servicio.leer_por_id(self.__pedir_entero("Id: ")))
                case _:
                    print("Opción inválida.")

    def __menu_precios(self) -> None:
        servicio = self.__contexto.precios
        while True:
            self.__titulo("Precios")
            opcion = self.__leer_opcion_crud()
            if opcion == "0":
                return
            match opcion:
                case "1":
                    self.__listar(servicio.leer_todos())
                case "2":
                    creado = servicio.crear(
                        Precio(
                            id=0,
                            libro=self.__exigir(
                                self.__contexto.libros.leer_por_id(
                                    self.__pedir_entero("Id de libro: ")
                                )
                            ),
                            moneda=self.__exigir(
                                self.__contexto.monedas.leer_por_id(
                                    self.__pedir_entero("Id de moneda: ")
                                )
                            ),
                            importe=self.__pedir_decimal("Importe: "),
                        )
                    )
                    print(f"Alta realizada: {creado}")
                case "3":
                    actual = self.__exigir(servicio.leer_por_id(self.__pedir_entero("Id: ")))
                    actual.importe = self.__pedir_decimal(
                        f"Importe [{actual.importe:.2f}]: ",
                        actual.importe,
                    )
                    print(f"Modificación realizada: {servicio.actualizar(actual)}")
                case "4":
                    self.__avisar_borrado(servicio.eliminar(self.__pedir_entero("Id: ")))
                case "5":
                    self.__mostrar(servicio.leer_por_id(self.__pedir_entero("Id: ")))
                case _:
                    print("Opción inválida.")

    def __menu_stock(self) -> None:
        servicio = self.__contexto.stocks
        while True:
            self.__titulo("Stock")
            opcion = self.__leer_opcion_crud()
            if opcion == "0":
                return
            match opcion:
                case "1":
                    self.__listar(servicio.leer_todos())
                case "2":
                    creado = servicio.crear(
                        Stock(
                            libro=self.__exigir(
                                self.__contexto.libros.leer_por_id(
                                    self.__pedir_entero("Id de libro: ")
                                )
                            ),
                            cantidad=self.__pedir_entero("Cantidad: "),
                            punto_reposicion=self.__pedir_entero("Punto de reposición: "),
                        )
                    )
                    print(f"Alta realizada: {creado}")
                case "3":
                    libro_id = self.__pedir_entero("Id de libro: ")
                    actual = self.__exigir(servicio.leer_por_libro(libro_id))
                    actual.cantidad = self.__pedir_entero(
                        f"Cantidad [{actual.cantidad}]: ", actual.cantidad
                    )
                    actual.punto_reposicion = self.__pedir_entero(
                        f"Punto de reposición [{actual.punto_reposicion}]: ",
                        actual.punto_reposicion,
                    )
                    print(f"Modificación realizada: {servicio.actualizar(actual)}")
                case "4":
                    self.__avisar_borrado(
                        servicio.eliminar(self.__pedir_entero("Id de libro: "))
                    )
                case "5":
                    self.__mostrar(
                        servicio.leer_por_libro(self.__pedir_entero("Id de libro: "))
                    )
                case _:
                    print("Opción inválida.")

    def __menu_cotizaciones(self) -> None:
        servicio = self.__contexto.cotizaciones
        while True:
            self.__titulo("Cotizaciones del dólar", "Lectura por tipo y fecha")
            opcion = self.__leer_opcion_crud()
            if opcion == "0":
                return
            match opcion:
                case "1":
                    self.__listar(servicio.leer_todas())
                case "2":
                    creado = servicio.crear(
                        CotizacionDolar(
                            tipo_cotizacion=self.__exigir(
                                self.__contexto.tipos_cotizacion.leer_por_id(
                                    self.__pedir_entero("Id de tipo: ")
                                )
                            ),
                            fecha=self.__pedir_fecha("Fecha AAAA-MM-DD: "),
                            compra=self.__pedir_decimal("Compra: "),
                            venta=self.__pedir_decimal("Venta: "),
                        )
                    )
                    print(f"Alta realizada: {creado}")
                case "3":
                    tipo_id = self.__pedir_entero("Id de tipo: ")
                    fecha = self.__pedir_fecha("Fecha AAAA-MM-DD: ")
                    actual = self.__exigir(servicio.leer_por_tipo_y_fecha(tipo_id, fecha))
                    actual.compra = self.__pedir_decimal(
                        f"Compra [{actual.compra:.2f}]: ", actual.compra
                    )
                    actual.venta = self.__pedir_decimal(
                        f"Venta [{actual.venta:.2f}]: ", actual.venta
                    )
                    print(f"Modificación realizada: {servicio.actualizar(actual)}")
                case "4":
                    self.__avisar_borrado(
                        servicio.eliminar(
                            self.__pedir_entero("Id de tipo: "),
                            self.__pedir_fecha("Fecha AAAA-MM-DD: "),
                        )
                    )
                case "5":
                    self.__mostrar(
                        servicio.leer_por_tipo_y_fecha(
                            self.__pedir_entero("Id de tipo: "),
                            self.__pedir_fecha("Fecha AAAA-MM-DD: "),
                        )
                    )
                case _:
                    print("Opción inválida.")

    def __menu_reportes(self) -> None:
        while True:
            print("\n--- Reportes ---")
            print("1. Comparación con Cúspide según un tipo de dólar")
            print("2. Stock a reponer")
            print("3. Histórico de cotización")
            print("4. Listado")
            print("5. Alta")
            print("6. Modificación")
            print("7. Borrado")
            print("8. Lectura")
            print("0. Volver")
            opcion = input("Opción: ").strip()
            if opcion == "0":
                return
            match opcion:
                case "1":
                    tipo_id = self.__pedir_entero("Id de tipo de cotización: ")
                    self.mostrar_comparacion(
                        self.__contexto.reportes.comparar_con_cuspide(tipo_id)
                    )
                case "2":
                    self.mostrar_stock(self.__contexto.reportes.stock_a_reponer())
                case "3":
                    tipo_id = self.__pedir_entero("Id de tipo de cotización: ")
                    self.mostrar_historico(self.__contexto.reportes.historico(tipo_id))
                case "4":
                    self.__listar_informes()
                case "5":
                    self.__alta_informe()
                case "6":
                    self.__modificar_informe()
                case "7":
                    self.__borrar_informe()
                case "8":
                    self.__leer_informe()
                case _:
                    print("Opción inválida.")

    def __listar_informes(self) -> None:
        if not self.__informes:
            print("No hay informes guardados.")
            return
        for identificador in sorted(self.__informes):
            informe = self.__informes[identificador]
            print(f"  [{informe.id}] {informe.titulo}")

    def __alta_informe(self) -> None:
        titulo = self.__pedir_texto("Título del informe: ")
        print("1. Comparación con Cúspide")
        print("2. Stock a reponer")
        print("3. Histórico de cotización")
        origen = input("Origen: ").strip()
        match origen:
            case "1":
                tipo_id = self.__pedir_entero("Id de tipo de cotización: ")
                detalle = self._texto_comparacion(
                    self.__contexto.reportes.comparar_con_cuspide(tipo_id)
                )
            case "2":
                detalle = self._texto_stock(self.__contexto.reportes.stock_a_reponer())
            case "3":
                tipo_id = self.__pedir_entero("Id de tipo de cotización: ")
                detalle = self._texto_historico(self.__contexto.reportes.historico(tipo_id))
            case _:
                raise ValueError("El origen del informe no es válido.")
        informe = _InformeGuardado(self.__proximo_informe, titulo, detalle)
        self.__informes[informe.id] = informe
        self.__proximo_informe += 1
        print(f"Alta realizada: [{informe.id}] {informe.titulo}")
        print(informe.detalle)

    def __modificar_informe(self) -> None:
        informe = self.__informes.get(self.__pedir_entero("Id: "))
        if informe is None:
            raise ValueError("No se encontró el informe.")
        informe.titulo = self.__pedir_texto(
            f"Título [{informe.titulo}]: ",
            informe.titulo,
        )
        print(f"Modificación realizada: [{informe.id}] {informe.titulo}")

    def __borrar_informe(self) -> None:
        identificador = self.__pedir_entero("Id: ")
        if identificador not in self.__informes:
            print("No se encontró el informe.")
            return
        del self.__informes[identificador]
        print("Informe eliminado.")

    def __leer_informe(self) -> None:
        informe = self.__informes.get(self.__pedir_entero("Id: "))
        if informe is None:
            print("No se encontró el informe.")
            return
        print(f"[{informe.id}] {informe.titulo}")
        print(informe.detalle)

    def mostrar_comparacion(self, filas: Sequence[ComparacionCuspide]) -> None:
        """Imprime la comparación contra el precio público de Cúspide."""
        print(self._texto_comparacion(filas))

    def mostrar_stock(self, filas: Sequence[LineaStock]) -> None:
        """Imprime los libros que hay que reponer."""
        print(self._texto_stock(filas))

    def mostrar_bimonetaria(self, filas: Sequence[LineaBimonetaria]) -> None:
        """Imprime el precio en pesos y en la otra moneda elegida."""
        if not filas:
            print("No hay libros para mostrar.")
            return
        for fila in filas:
            pesos = (
                formatear_pesos(fila.precio_ars)
                if fila.precio_ars is not None
                else "sin precio"
            )
            otra = (
                f"{fila.precio_otra:.2f} {fila.codigo_moneda}"
                if fila.precio_otra is not None
                else f"sin precio {fila.codigo_moneda}"
            )
            print(f"  {fila.titulo} ({fila.autor}) | ARS {pesos} | {otra}")

    def __ejecutar_scraping(self) -> None:
        """Pide la diferencia y corre el scraper antes de armar las alertas."""
        umbral = self.__pedir_decimal("Diferencia de precios para alertar: ")
        print(f"Umbral de diferencia ingresado: {umbral}")
        ruta = self.__contexto.scraping.ejecutar(umbral)
        print(f"Alertas generadas en {ruta}")

    def __ver_auditoria(self) -> None:
        """Imprime el historial de auditoría."""
        historial = self.__contexto.auditoria.leer_todos()
        if not historial:
            print("No hay auditorías registradas.")
            return
        for fila in historial:
            print(f"  {fila}")

    def __obtener_cotizaciones_api(self) -> None:
        """Pide las cotizaciones a la API y sigue si la red falla."""
        try:
            registradas = self.__contexto.cotizaciones.obtener_cotizaciones()
        except (ValueError, URLError, OSError) as error:
            print(f"No se pudo obtener las cotizaciones: {error}")
            return
        if not registradas:
            print("La API no devolvió cotizaciones para registrar.")
            return
        print("Cotizaciones registradas desde la API:")
        for cotizacion in registradas:
            print(f"  {cotizacion}")

    def __lista_bimonetaria(self) -> None:
        print("Monedas:")
        self.__listar(self.__contexto.monedas.leer_todos())
        codigo = self.__pedir_texto("Código de la segunda moneda: ")
        self.mostrar_bimonetaria(self.__contexto.reportes.lista_bimonetaria(codigo))

    def mostrar_historico(self, filas: Sequence[LineaHistorico]) -> None:
        """Imprime el histórico de un tipo de dólar."""
        print(self._texto_historico(filas))

    def _texto_comparacion(self, filas: Sequence[ComparacionCuspide]) -> str:
        """Arma el texto de la comparación con Cúspide."""
        if not filas:
            return "No hay libros con precio en pesos y en dólares."
        lineas = [
            f"Cotización usada: {filas[0].tipo_cotizacion} "
            f"venta {formatear_pesos(filas[0].cotizacion_venta)}"
        ]
        for fila in filas:
            estado = "cubre el costo" if fila.cubre_el_costo else "no cubre el costo"
            lineas.append(
                f"  {fila.titulo} ({fila.autor}) | "
                f"Cúspide {formatear_pesos(fila.precio_cuspide_ars)} | "
                f"costo USD {fila.costo_usd:.2f} -> "
                f"{formatear_pesos(fila.costo_reposicion_ars)} | "
                f"diferencia {formatear_pesos(fila.diferencia_ars)} ({estado})"
            )
        return "\n".join(lineas)

    def _texto_stock(self, filas: Sequence[LineaStock]) -> str:
        """Arma el texto del stock a reponer."""
        if not filas:
            return "No hay libros en el punto de reposición."
        lineas = []
        for fila in filas:
            texto = "1 unidad" if fila.cantidad == 1 else f"{fila.cantidad} unidades"
            lineas.append(
                f"  {fila.titulo} ({fila.autor}): "
                f"{texto}, reponer al llegar a {fila.punto_reposicion}"
            )
        return "\n".join(lineas)

    def _texto_historico(self, filas: Sequence[LineaHistorico]) -> str:
        """Arma el texto del histórico de cotización."""
        if not filas:
            return "No hay cotizaciones para ese tipo."
        return "\n".join(
            f"  {fila.fecha.isoformat()} {fila.tipo}: "
            f"compra {formatear_pesos(fila.compra)} / "
            f"venta {formatear_pesos(fila.venta)}"
            for fila in filas
        )

    def __titulo(self, texto: str, lectura: str = "Lectura") -> None:
        print(f"\n--- {texto} ---")
        print(
            "1. Listado  2. Alta  3. Modificación  4. Borrado  "
            f"5. {lectura}  0. Volver"
        )

    def __leer_opcion_crud(self) -> str:
        return input("Opción: ").strip()

    def __listar(self, entidades: Sequence[object]) -> None:
        if not entidades:
            print("No hay registros.")
            return
        for entidad in entidades:
            print(f"  {entidad}")

    def __mostrar(self, entidad: Optional[object]) -> None:
        if entidad is None:
            print("No se encontró el registro.")
            return
        print(entidad)

    def __exigir(self, entidad: Optional[object]) -> object:
        if entidad is None:
            raise ValueError("No se encontró el registro.")
        return entidad

    def __avisar_borrado(self, eliminado: bool) -> None:
        if eliminado:
            print("Registro eliminado.")
            return
        print("No se encontró el registro.")

    def __pedir_texto(self, mensaje: str, valor_actual: Optional[str] = None) -> str:
        texto = input(mensaje).strip()
        if texto == "" and valor_actual is not None:
            return valor_actual
        if texto == "":
            raise ValueError("El texto no puede estar vacío.")
        return texto

    def __pedir_entero(
        self,
        mensaje: str,
        valor_actual: Optional[int] = None,
    ) -> int:
        texto = input(mensaje).strip()
        if texto == "" and valor_actual is not None:
            return valor_actual
        if not texto.isdigit():
            raise ValueError("Se esperaba un número entero mayor o igual a cero.")
        return int(texto)

    def __pedir_decimal(
        self,
        mensaje: str,
        valor_actual: Optional[float] = None,
    ) -> float:
        texto = input(mensaje).strip().replace(",", ".")
        if texto == "" and valor_actual is not None:
            return valor_actual
        try:
            return float(texto)
        except ValueError as error:
            raise ValueError("Se esperaba un número.") from error

    def __pedir_fecha(self, mensaje: str) -> date:
        texto = input(mensaje).strip()
        try:
            return date.fromisoformat(texto)
        except ValueError as error:
            raise ValueError("La fecha debe tener el formato AAAA-MM-DD.") from error


def ejecutar_demostracion(contexto: Contexto) -> None:
    """Lista cada modelo y hace un alta o una modificación, y luego los reportes.

    La segunda ejecución modifica el registro de demostración en lugar de
    volver a darlo de alta.

    Args:
        contexto (Contexto): Servicios ya armados.
    """
    consola = Consola(contexto)
    print("Book Manager - demostración del Sprint 3")
    print("Catálogo de referencia: https://cuspide.com/")
    print("Grupo 28")
    _demostrar_genero(contexto)
    _demostrar_editorial(contexto)
    _demostrar_moneda(contexto)
    _demostrar_tipo(contexto)
    libro = _demostrar_libro(contexto)
    _demostrar_precio(contexto, libro.id)
    _demostrar_stock(contexto, libro.id)
    _demostrar_cotizacion(contexto)
    print("\n=== Reportes ===")
    print("\nComparación con Cúspide al dólar blue")
    blue = contexto.tipos_cotizacion.buscar_por_nombre("Blue")
    if blue is None:
        raise ValueError("Falta el tipo de cotización Blue en los datos iniciales.")
    consola.mostrar_comparacion(contexto.reportes.comparar_con_cuspide(blue.id))
    print("\nComparación con Cúspide al dólar oficial")
    oficial = contexto.tipos_cotizacion.buscar_por_nombre("Oficial")
    if oficial is None:
        raise ValueError("Falta el tipo de cotización Oficial en los datos iniciales.")
    consola.mostrar_comparacion(contexto.reportes.comparar_con_cuspide(oficial.id))
    print("\nStock a reponer")
    consola.mostrar_stock(contexto.reportes.stock_a_reponer())
    print("\nHistórico del dólar blue")
    consola.mostrar_historico(contexto.reportes.historico(blue.id))
    print("\n=== 10. Obtener las cotizaciones por API ===")
    try:
        cotizaciones_api = contexto.cotizaciones.obtener_cotizaciones()
        if not cotizaciones_api:
            print("  La API no devolvió cotizaciones para registrar.")
        for cotizacion in cotizaciones_api:
            print(f"  {cotizacion}")
    except (ValueError, URLError, OSError) as error:
        print(f"  No se pudo consultar la API: {error}")
    print("\n=== 11. Lista de precios bimonetaria ===")
    try:
        consola.mostrar_bimonetaria(contexto.reportes.lista_bimonetaria("USD"))
    except (ValueError, OSError) as error:
        print(f"  No se pudo armar la lista bimonetaria: {error}")
    print("\n=== 12. Exportar los precios a CSV ===")
    try:
        ruta = contexto.reportes.exportar_precios_csv()
        print(f"  Precios exportados en {ruta}")
    except (ValueError, OSError) as error:
        print(f"  No se pudo exportar el CSV: {error}")
    print("\n=== Carga de datos desde SQL ===")
    try:
        from book_manager.migrations.migrations import cargar_datos_desde_sql

        carpeta_sql = Path(__file__).resolve().parents[1] / "migrations" / "sql"
        for archivo, cantidad in cargar_datos_desde_sql(str(carpeta_sql)).items():
            print(f"  {archivo}: {cantidad} filas")
    except (FileNotFoundError, ValueError, OSError) as error:
        print(f"  No se pudo cargar desde SQL: {error}")
    print("\n=== 13. Ejecutar scraping ===")
    umbral = 0.0
    print(f"  Umbral de diferencia ingresado: {umbral}")
    try:
        ruta_alertas = contexto.scraping.ejecutar(umbral)
        print(f"  Alertas CSV: {ruta_alertas}")
        with ruta_alertas.open(encoding="utf-8") as archivo:
            for linea in archivo.readlines()[:8]:
                print(f"  {linea.rstrip()}")
    except (ValueError, OSError) as error:
        print(f"  No se pudo ejecutar el scraping: {error}")
    print("\n=== 14. Generar reporte ===")
    try:
        ruta_excel = contexto.scraping.generar_reporte_excel()
        print(f"  Reporte Excel: {ruta_excel}")
    except (ValueError, OSError) as error:
        print(f"  No se pudo generar el Excel: {error}")
    print("\n=== 15. Ver historial de auditoría ===")
    try:
        historial = contexto.auditoria.leer_todos()
        print(f"  Registros: {len(historial)}")
        for fila in historial[-8:]:
            print(f"  {fila}")
    except (ValueError, OSError) as error:
        print(f"  No se pudo leer la auditoría: {error}")
    print("\nDemostración finalizada.")


def _demostrar_genero(contexto: Contexto) -> None:
    print("\n--- Géneros: listado ---")
    _imprimir(contexto.generos.leer_todos())
    existente = contexto.generos.buscar_por_nombre("Teatro")
    if existente is None:
        creado = contexto.generos.crear(
            Genero(
                id=0,
                nombre="Teatro",
                descripcion="Género cargado en la demostración del sprint 1.",
            )
        )
        print(f"Alta realizada: {creado}")
        return
    existente.descripcion = "Género actualizado en la demostración del sprint 1."
    print(f"Modificación realizada: {contexto.generos.actualizar(existente)}")


def _demostrar_editorial(contexto: Contexto) -> None:
    print("\n--- Editoriales: listado ---")
    _imprimir(contexto.editoriales.leer_todos())
    existente = contexto.editoriales.buscar_por_nombre("Cátedra")
    if existente is None:
        creado = contexto.editoriales.crear(
            Editorial(
                id=0,
                nombre="Cátedra",
                pais="España",
                sitio_web="https://www.catedra.com",
            )
        )
        print(f"Alta realizada: {creado}")
        return
    existente.sitio_web = "https://www.catedra.com"
    print(f"Modificación realizada: {contexto.editoriales.actualizar(existente)}")


def _demostrar_moneda(contexto: Contexto) -> None:
    print("\n--- Monedas: listado ---")
    _imprimir(contexto.monedas.leer_todos())
    existente = contexto.monedas.buscar_por_codigo("XXX")
    if existente is None:
        creado = contexto.monedas.crear(
            Moneda(id=0, codigo="XXX", nombre="Moneda de prueba", simbolo="X")
        )
        print(f"Alta realizada: {creado}")
        return
    existente.nombre = "Moneda de prueba del sprint 1"
    print(f"Modificación realizada: {contexto.monedas.actualizar(existente)}")


def _demostrar_tipo(contexto: Contexto) -> None:
    print("\n--- Tipos de cotización: listado ---")
    _imprimir(contexto.tipos_cotizacion.leer_todos())
    existente = contexto.tipos_cotizacion.buscar_por_nombre("Prueba")
    if existente is None:
        creado = contexto.tipos_cotizacion.crear(
            TipoCotizacion(
                id=0,
                nombre="Prueba",
                descripcion="Tipo cargado en la demostración del sprint 1.",
            )
        )
        print(f"Alta realizada: {creado}")
        return
    existente.descripcion = "Tipo actualizado en la demostración del sprint 1."
    print(f"Modificación realizada: {contexto.tipos_cotizacion.actualizar(existente)}")


def _demostrar_libro(contexto: Contexto) -> Libro:
    print("\n--- Libros: listado ---")
    _imprimir(contexto.libros.leer_todos())
    isbn = "978-950-28-0099-1"
    existente = contexto.libros.buscar_por_isbn(isbn)
    if existente is None:
        creado = contexto.libros.crear(
            Libro(
                id=0,
                isbn=isbn,
                titulo="Libro de demostración del sprint 1",
                autor="Grupo 28",
                editorial=contexto.editoriales.leer_por_id(1),
                genero=contexto.generos.leer_por_id(1),
                anio=2026,
                paginas=128,
            )
        )
        print(f"Alta realizada: {creado}")
        return creado
    existente.paginas = 128
    actualizado = contexto.libros.actualizar(existente)
    print(f"Modificación realizada: {actualizado}")
    return actualizado


def _demostrar_precio(contexto: Contexto, libro_id: int) -> None:
    print("\n--- Precios: listado ---")
    _imprimir(contexto.precios.leer_todos())
    moneda = contexto.monedas.buscar_por_codigo("ARS")
    if moneda is None:
        raise ValueError("Falta la moneda ARS en los datos iniciales.")
    existente = contexto.precios.buscar(libro_id, moneda.id)
    if existente is None:
        creado = contexto.precios.crear(
            Precio(
                id=0,
                libro=contexto.libros.leer_por_id(libro_id),
                moneda=moneda,
                importe=15000,
            )
        )
        print(f"Alta realizada: {creado}")
        return
    existente.importe = 15000
    print(f"Modificación realizada: {contexto.precios.actualizar(existente)}")


def _demostrar_stock(contexto: Contexto, libro_id: int) -> None:
    print("\n--- Stock: listado ---")
    _imprimir(contexto.stocks.leer_todos())
    existente = contexto.stocks.leer_por_libro(libro_id)
    if existente is None:
        creado = contexto.stocks.crear(
            Stock(
                libro=contexto.libros.leer_por_id(libro_id),
                cantidad=4,
                punto_reposicion=2,
            )
        )
        print(f"Alta realizada: {creado}")
        return
    existente.cantidad = 4
    print(f"Modificación realizada: {contexto.stocks.actualizar(existente)}")


def _demostrar_cotizacion(contexto: Contexto) -> None:
    print("\n--- Cotizaciones del dólar: listado ---")
    _imprimir(contexto.cotizaciones.leer_todas())
    oficial = contexto.tipos_cotizacion.buscar_por_nombre("Oficial")
    if oficial is None:
        raise ValueError("Falta el tipo de cotización Oficial en los datos iniciales.")
    fecha = date(2026, 10, 2)
    existente = contexto.cotizaciones.leer_por_tipo_y_fecha(oficial.id, fecha)
    if existente is None:
        creado = contexto.cotizaciones.crear(
            CotizacionDolar(
                tipo_cotizacion=oficial,
                fecha=fecha,
                compra=1490,
                venta=1540,
            )
        )
        print(f"Alta realizada: {creado}")
        return
    existente.compra = 1490
    existente.venta = 1540
    print(f"Modificación realizada: {contexto.cotizaciones.actualizar(existente)}")


def _imprimir(entidades: Sequence[object]) -> None:
    if not entidades:
        print("  No hay registros.")
        return
    for entidad in entidades:
        print(f"  {entidad}")
''', encoding='utf-8')
print('Vista —', destino.name, destino.stat().st_size, 'bytes')
````

````python
#@title Ejercicio 07 — main
from pathlib import Path
destino = Path(RUTA_SRC) / 'book_manager/main.py'
destino.parent.mkdir(parents=True, exist_ok=True)
destino.write_text(r'''"""Punto de entrada de Book Manager."""

from pathlib import Path

from sqlalchemy import text

import book_manager.models.models  # registra las tablas en Base
from book_manager.database.connection import Base
from book_manager.database.connection import ConexionDB
from book_manager.migrations.migrations import migrar_datos
from book_manager.repositories.repositories import RepositorioAuditoria
from book_manager.repositories.repositories import RepositorioCotizacionDolar
from book_manager.repositories.repositories import RepositorioEditorial
from book_manager.repositories.repositories import RepositorioExtraccionWeb
from book_manager.repositories.repositories import RepositorioGenero
from book_manager.repositories.repositories import RepositorioLibro
from book_manager.repositories.repositories import RepositorioMoneda
from book_manager.repositories.repositories import RepositorioPrecio
from book_manager.repositories.repositories import RepositorioStock
from book_manager.repositories.repositories import RepositorioTipoCotizacion
from book_manager.services.services import Contexto
from book_manager.services.services import EditorialServicio
from book_manager.services.services import GeneroServicio
from book_manager.services.services import LibroServicio
from book_manager.services.services import MonedaServicio
from book_manager.services.services import PrecioServicio
from book_manager.services.services import ReporteServicio
from book_manager.services.services import ServicioAuditoria
from book_manager.services.services import ServicioCotizacionDolar
from book_manager.services.services import ServicioScraping
from book_manager.services.services import StockServicio
from book_manager.services.services import TipoCotizacionServicio
from book_manager.services.services import cargar_env
from book_manager.ui.console import Consola
from book_manager.ui.console import ejecutar_demostracion


def _base_vacia(db: ConexionDB) -> bool:
    """Indica si todavía no hay libros persistidos."""
    with db.engine.connect() as conexion:
        tablas = conexion.execute(
            text("SELECT name FROM sqlite_master WHERE type = 'table'")
        ).scalars()
        if "libros" not in set(tablas):
            return True
        cantidad = conexion.execute(text("SELECT COUNT(*) FROM libros")).scalar()
    return not cantidad


def preparar_base(importar_semillas: bool) -> ConexionDB:
    """Crea las tablas y, si hace falta, migra los CSV del Sprint 1.

    Args:
        importar_semillas (bool): Si es True, vacía la base y vuelve a migrar.

    Returns:
        ConexionDB: Conexión lista para los repositorios.
    """
    db = ConexionDB()
    if importar_semillas or _base_vacia(db):
        Base.metadata.drop_all(bind=db.engine)
        db.crear_tablas()
        paquete = Path(__file__).resolve().parent
        migrar_datos(
            str(paquete / "migrations" / "csv"),
            str(paquete / "migrations" / "sql"),
            db=db,
        )
        return db
    db.crear_tablas()
    return db


def construir_contexto(db: ConexionDB) -> Contexto:
    """Arma repositorios y servicios sobre la conexión indicada.

    Args:
        db (ConexionDB): Conexión compartida por todos los repositorios.

    Returns:
        Contexto: Servicios listos para la consola o la demostración.
    """
    generos = RepositorioGenero(db)
    editoriales = RepositorioEditorial(db)
    monedas = RepositorioMoneda(db)
    tipos = RepositorioTipoCotizacion(db)
    libros = RepositorioLibro(db)
    precios = RepositorioPrecio(db)
    stocks = RepositorioStock(db)
    cotizaciones = RepositorioCotizacionDolar(db)
    extracciones = RepositorioExtraccionWeb(db)
    auditorias = RepositorioAuditoria(db)
    return Contexto(
        generos=GeneroServicio(generos, libros),
        editoriales=EditorialServicio(editoriales, libros),
        monedas=MonedaServicio(monedas, precios),
        tipos_cotizacion=TipoCotizacionServicio(tipos, cotizaciones),
        libros=LibroServicio(libros, generos, editoriales, precios, stocks),
        precios=PrecioServicio(precios, libros, monedas),
        stocks=StockServicio(stocks, libros),
        cotizaciones=ServicioCotizacionDolar(cotizaciones, tipos),
        reportes=ReporteServicio(
            libros,
            precios,
            monedas,
            stocks,
            cotizaciones,
            tipos,
        ),
        scraping=ServicioScraping(extracciones, libros, precios),
        auditoria=ServicioAuditoria(auditorias),
    )


def main(import_default_data: bool = False, interactivo: bool = False) -> None:
    """Inicia el sistema sobre la base relacional.

    Con interactivo en False ejecuta el listado y el alta o la modificación
    de cada modelo, los reportes, la API, la lista bimonetaria y la exportación
    a CSV, y termina. Así el notebook puede continuar con las celdas de git.
    Con interactivo en True abre la consola.

    Args:
        import_default_data (bool): Si es True, vacía la base y migra los CSV.
        interactivo (bool): Si es True, abre el menú de consola.
    """
    cargar_env()
    db = preparar_base(importar_semillas=import_default_data)
    contexto = construir_contexto(db)
    if interactivo:
        Consola(contexto).ejecutar()
        return
    ejecutar_demostracion(contexto)


if __name__ == "__main__":
    main(import_default_data=False, interactivo=True)
''', encoding='utf-8')
print('Vista —', destino.name, destino.stat().st_size, 'bytes')
````
